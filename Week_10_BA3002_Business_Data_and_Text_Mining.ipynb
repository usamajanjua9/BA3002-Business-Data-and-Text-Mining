## 📄 **1. Challenges Associated with Unstructured Information**

### 🔹 **1.1 What is Unstructured Information?**

#### **Definition**
**Unstructured information** is information that does not follow a fixed tabular structure such as rows and columns.

Structured data usually fits naturally into a table:

| Customer_ID | Age | Rating | Amount |
|---|---:|---:|---:|
| C001 | 24 | 4 | 12,500 |
| C002 | 31 | 2 | 8,000 |

Unstructured information may appear as:

- Customer reviews ⭐
- Complaint messages 📩
- Emails 📧
- Survey responses 📝
- News articles 📰
- Social media posts 📱
- Chat conversations 💬
- Call transcripts 🎙️
- Business reports 📑

#### **Example**
> “The product is excellent, but delivery was extremely slow.”

A database can store:
- Rating = 3
- Date = 14-Oct
- Product = Headphones

But the reason behind the rating is hidden in the written comment.

#### 💡 **Key Idea**
**Numbers often tell us _what_ happened. Text often tells us _why_ it happened.**

#### 🚨 **Business Problem**
In this hypothetical example, an e-commerce company receives 50,000 reviews every month. Management wants to know:

- Why are customers unhappy?
- Which products receive the most complaints?
- Are complaints mainly about delivery, quality, price, or service?
- What features do customers appreciate?
- What should be improved first?

Reading every review manually is not practical.

#### ✅ **Solution**
Business Text Mining can automatically discover patterns:

**Process:** 50,000 Customer Reviews → Text Mining → Repeated Themes → Delivery Delay → Damaged Product → Poor Support → Pricing → Business Decision.

The objective is not simply to count words. The objective is to transform text into **actionable business knowledge**.

#### **Structured, Semi-Structured, and Unstructured Data**

| Form | Business example | What is organized? |
|---|---|---|
| Structured | Rating, product ID, transaction amount | Fixed fields and expected types |
| Semi-structured | JSON API response or HTML review page | Keys, nesting, or tags provide a framework |
| Unstructured | Free-text review inside the JSON response | Meaning is not already separated into analytical fields |

Putting a review in a spreadsheet does not make its meaning structured. A single record can contain structured metadata and unstructured text. Text describes the customer's reported experience; it does not independently prove the cause.

### 🔹 **1.2 Different Words, Similar Meaning**

#### **Definition**
Human language is flexible, contextual, and inconsistent. People can express the **same idea using different words**.

#### **Example**
1. “Delivery was late.”
2. “The package arrived after a week.”
3. “Shipping took forever.”
4. “Very slow delivery.”
5. “My order did not arrive on time.”

All describe approximately the same issue:

> **Delivery Delay**

#### 🚨 **Problem**
A simple keyword rule looking only for **late** may detect only one comment.

#### ✅ **Solution**
NLP techniques can use related words, context, word combinations, and learned patterns.

**Process:** "late delivery" → "shipping took forever" → "did not arrive on time" → Similar Meaning → DELIVERY DELAY.

### 🔹 **1.3 Ambiguity in Human Language**

#### **Definition**
**Ambiguity** occurs when the same word or sentence can have more than one meaning.

#### **Example**
> “The bank is closed.”

**Bank** may refer to:
- 🏦 A financial institution
- 🌊 The side of a river

Another example:

> “There is a charge on my account.”  
> “My phone does not charge.”

The same word has different meanings.

#### 🚨 **Problem**
If every occurrence of **charge** is treated as a financial complaint, the results become inaccurate.

#### ✅ **Solution**
NLP considers surrounding words and context.

**Process:** "bank charged an extra fee" → Financial Complaint → "phone does not charge" → Technical Product Issue.

#### 💡 **Business Lesson**
A word by itself may not be enough. **Context creates meaning.**

### 🔹 **1.4 Noisy Customer-Generated Text**

#### **Definition**
**Noisy text** contains spelling mistakes, abbreviations, emojis, repeated characters, URLs, inconsistent capitalization, slang, or unnecessary symbols.

#### **Example**
```text
OMG!!! 😡😡 delivery waz sooo LATE!!!
www.shop.com/order/29292
NEVER buying again!!!
```

A human understands the complaint immediately. A computer initially sees many inconsistent elements.

#### 🚨 **Problem**
These may appear as different forms:

```text
good
Good
GOOD
goooood
good!!!
```

#### ✅ **Solution**
Text preprocessing can standardize textual information.

**Process:** Original: → "OMG!!! Delivery waz sooo LATE 😡" → Cleaned: → "delivery was late".

Common operations include lowercasing, URL removal, punctuation handling, tokenization, and lemmatization.

> These techniques are developed in detail in **Week 11**.

> **Keep the signal:** Emojis, repeated letters, and punctuation can express emotion. Decide what to preserve before removing them.

### 🔹 **1.5 Context Changes Meaning**

#### **Definition**
The meaning of a word often depends on the words around it.

#### **Example**
**Banking:** “The bank charged me an extra fee.”  
**Technology:** “My phone does not charge.”  
**Legal:** “The company faced a fraud charge.”

#### 🚨 **Problem**
Counting **charge** without context mixes different business situations.

#### ✅ **Solution**
NLP examines surrounding words and patterns.

**Process:** Word + Context → Meaning → Business Category.

### 🔹 **1.6 Negation Changes Meaning**

#### **Definition**
**Negation** occurs when words such as **not, never, no, isn't, wasn't, don't** reverse or modify meaning.

#### **Example**
> “The service is good.”

versus:

> “The service is **not good**.”

#### 🚨 **Problem**
Careless preprocessing can remove the word **not**:

```text
Original:
"The service is not good."

Incorrect Processing:
"service good"
```

The meaning changes completely.

#### ✅ **Solution**
Important negation terms should be preserved when they affect meaning.

**Process:** "service is not good" → Negative Service Experience.

#### 💡 **Business Lesson**
Text preprocessing should support the **business objective**, not blindly remove words.

### 🔹 **1.7 Sarcasm and Hidden Meaning**

#### **Definition**
**Sarcasm** occurs when words appear positive but actually communicate criticism.

#### **Example**
> “Amazing service! Only took them three weeks to deliver my order.”

> “Great. The app crashed again.”

#### 🚨 **Problem**
A simple sentiment rule may see:

```text
Amazing → Positive
Great   → Positive
```

and reach the wrong conclusion.

#### ✅ **Solution**
More advanced NLP systems examine sentence context, word combinations, punctuation, contradictions, and contextual language representations.

#### 💡 **Key Insight**
**Positive words do not always mean positive sentiment.**

> Context-aware models can still misinterpret sarcasm. Review uncertain cases and measure errors on representative examples.

### 🔹 **1.8 Multilingual and Mixed-Language Business Text**

#### **Definition**
Business text may contain multiple languages, sometimes within the same sentence.

#### **Example**
> “Product acha hai but delivery bohat slow thi.”

> “App ka interface good hai lekin payment issue aa raha hai.”

#### 🚨 **Problem**
A system designed only for standard English may fail to interpret words such as:

- acha
- bohat
- lekin
- aa raha hai

#### ✅ **Solution**
Possible approaches include:

- Language detection
- Multilingual NLP models
- Translation
- Domain-specific dictionaries
- Multilingual embeddings
- Custom preprocessing

#### 💡 **Business Lesson**
Language diversity affects whether a business can **hear the full voice of its customers**.

### 🔹 **1.9 Scale and Volume of Textual Information**

#### **Definition**
**Scale** refers to the large amount of text continuously produced across digital channels.

#### **Example**
An illustrative platform receives every day:

- 20,000 product reviews
- 15,000 support messages
- 10,000 social media mentions
- 8,000 open-ended survey responses

```text
53,000 text records per day
= 1.59 million records in a 30-day month
```

#### 🚨 **Problem**
Manual analysis becomes slow, expensive, inconsistent, and difficult to scale.

#### ✅ **Solution**
Automated text mining can process large volumes consistently.

**Process:** Millions of Text Records → Automated Text Processing → Topics + Patterns + Sentiment → Business Insights.

### 🔹 **1.10 Human vs Machine: Interpret the Same Review**

**Definition**  
A useful interpretation separates the topic, opinion, evidence, and uncertainty.

**Example**  
> “Great. Another payment failure. Product acha hai, app nahi.”

**🚨 Problem**  
A positive-word counter may report satisfaction; an English-only system may miss the Roman Urdu criticism.

**✅ Solution**  
Identify payment reliability as a negative aspect, treat “Great” as potentially sarcastic, and request language-aware review. Do not assume the product itself is disliked.

**Discuss:** What can you infer directly? What would require a follow-up question or an operational log?

## 🌍 **2. Real-World Examples from E-Commerce, Banking, Hospitality, and Retail**

### 🔹 **2.1 E-Commerce**

#### **Definition**
E-commerce companies generate text through product reviews, seller reviews, questions, return reasons, complaints, support conversations, and social media.

#### **Example**
> “Sound quality is excellent.”  
> “Battery dies after two hours.”  
> “Very comfortable but microphone quality is poor.”  
> “Good product but overpriced.”

#### 🚨 **Business Problem**
Management sees an average rating of **3.8 / 5**, but this number does not explain **why** customers gave those ratings.

#### ✅ **Solution**
Text mining can separate feedback by aspect:

| Aspect | Typical Opinion |
|---|---|
| 🎵 Sound | Positive |
| 🔋 Battery | Negative |
| 🎧 Comfort | Positive |
| 🎤 Microphone | Negative |
| 💰 Price | Mixed |

#### **Business Decision**
- Improve battery performance
- Redesign the microphone
- Review pricing
- Preserve strong sound quality

> **Ratings show how customers scored the product. Text explains why.**

### 🔹 **2.2 Banking and FinTech**

#### **Definition**
Banks and FinTech organizations receive text through complaints, emails, chatbots, call transcripts, app reviews, fraud reports, surveys, and social media.

#### **Example**
> “My transfer is still pending.”  
> “I was charged twice for one transaction.”  
> “The app keeps crashing during payment.”  
> “Customer support never replied.”

#### 🚨 **Business Problem**
Management needs to determine the biggest recurring issues across thousands of complaints.

#### ✅ **Solution**
Text mining can group complaints:

| Complaint Topic | Interpretation |
|---|---|
| 💸 Failed Transactions | Payment or transfer failed |
| ⏳ Delayed Transfers | Transaction remains pending |
| 📱 Application Issues | App crashes or freezes |
| ☎️ Customer Support | Slow or missing response |
| 💳 Card Problems | Card blocked, declined, or lost |

#### **Business Decision**
Resources can be prioritized according to the most frequent and damaging customer problems.

### 🔹 **2.3 Hospitality and Tourism**

#### **Definition**
Hotels, restaurants, airlines, and tourism businesses rely heavily on written customer feedback.

#### **Example**
> “Excellent location near the city center.”  
> “Room was clean but breakfast was disappointing.”  
> “Staff were very friendly.”  
> “Wi-Fi was extremely slow.”  
> “The room was too noisy at night.”

#### 🚨 **Business Problem**
An overall rating does not identify which service areas need attention.

#### ✅ **Solution**
Text can be grouped by service feature:

| Feature | Typical Sentiment |
|---|---|
| 📍 Location | Positive |
| 🧹 Cleanliness | Positive |
| 👨‍💼 Staff | Positive |
| 🍳 Breakfast | Negative |
| 📶 Wi-Fi | Negative |
| 🔊 Noise | Negative |

#### **Business Decision**
Improve weak service areas while preserving what customers already value.

### 🔹 **2.4 Retail**

#### **Definition**
Retail businesses collect text from feedback forms, surveys, reviews, complaint emails, support conversations, return explanations, and social media.

#### **Example**
> “Checkout line was too long.”  
> “Staff were very helpful.”  
> “The product I wanted was out of stock.”  
> “Prices are higher than competitors.”

#### 🚨 **Business Problem**
Customer satisfaction may fall without a clear explanation from numerical scores alone.

#### ✅ **Solution**
Text mining may reveal recurring themes:

**Process:** Customer Feedback → Text Mining → Top Issues → Long checkout queues → Product availability → Pricing → Store navigation.

#### **Business Decision**
- Open additional checkout counters
- Improve inventory forecasting
- Review prices
- Improve store layout

### 🔹 **2.5 One Review, Multiple Business Aspects**

#### **Definition**
A single review can discuss multiple **aspects** of a product or service, with a different opinion about each aspect.

#### **Example**
> “The laptop performs really well, but delivery took ten days and the packaging was damaged.”

| Aspect | Customer Opinion |
|---|---|
| Laptop Performance | ✅ Positive |
| Delivery | ❌ Negative |
| Packaging | ❌ Negative |

#### 🚨 **Problem**
If the whole review receives one overall negative label, management may wrongly assume the customer disliked the laptop itself.

#### ✅ **Solution**
Analyze individual aspects:

**Process:** Review → Product Performance → Positive → Delivery → Negative → Packaging → Negative.

#### 💡 **Business Lesson**
Ask **what exactly** the customer is positive or negative about.

## 🧠 **3. Introduction to Natural Language Processing**

### 🔹 **3.1 What is NLP?**

#### **Definition**
**Natural Language Processing (NLP)** is a field of Artificial Intelligence that enables computers to **process, analyze, understand, and generate human language**.

#### **Example**
Customer message:

> “My order still hasn't arrived.”

Possible interpretation:

```text
Intent    → Complaint
Topic     → Delivery
Sentiment → Negative
Urgency   → High
```

#### 🚨 **Problem**
Computers easily process values such as:

```text
Revenue = 120000
Rating = 4
Quantity = 15
```

But this requires interpretation:

> “The product is decent, although I expected better.”

#### ✅ **Solution**
NLP transforms language into machine-processable representations.

**Process:** Human Language → Text Processing → Language Representation → NLP Model → Meaning / Pattern → Business Insight.

### 🔹 **3.2 How Computers Process Human Language**

#### **Example**
> “The delivery was not good.”

#### 👨 **Human Interpretation**
- Topic → Delivery
- Opinion → Negative
- Business area → Logistics / Service

#### 🚨 **Problem**
Human language contains meaning, context, emotion, grammar, relationships, and intent. These are not directly numerical.

#### ✅ **Solution**
NLP creates machine-processable representations:

**Process:** Raw Review → "The delivery was not good." → ["delivery", "not", "good"] → Language Representation → Negative Delivery Experience.

### 🔹 **3.3 Natural Language Understanding (NLU)**

#### **Definition**
**NLU** focuses on determining the **meaning, intent, entities, and context** in human language.

#### **Example**
> “Please block my card. I lost my wallet yesterday.”

```text
Intent  → Block Card
Entity  → Card
Reason  → Lost Wallet
Urgency → High
```

#### 🚨 **Problem**
Customers may write:
- “Block my card.”
- “Please deactivate my debit card.”
- “Someone stole my wallet.”
- “I don't have my card anymore.”

#### ✅ **Solution**
NLU focuses on **meaning rather than exact wording**.

**Process:** Different Customer Messages → Understanding Intent → CARD SECURITY REQUEST.

### 🔹 **3.4 Natural Language Generation (NLG)**

#### **Definition**
**NLG** enables computers to produce human-readable language from structured data or internal information.

#### **Example**
System data:

```text
Transaction Status = Failed
Amount = PKR 10,000
Refund Time = 3 business days
```

Generated message:

> “Your PKR 10,000 transaction was unsuccessful. The amount will be returned within three business days.”

#### 🚨 **Problem**
Raw system codes can be difficult for customers to understand.

#### ✅ **Solution**
**Process:** Structured Data → NLG → Human-Readable Message.

### 🔹 **3.5 NLU vs NLG**

| | **NLU 🧠** | **NLG ✍️** |
|---|---|---|
| Main Purpose | Interpret language | Generate language |
| Input | Human language | Data / context |
| Output | Meaning or intent | Human-readable text |
| Banking Example | Detect a lost-card request | Generate confirmation |
| E-Commerce Example | Detect a return request | Generate return instructions |

```text
Customer → Text → NLU → Computer Understands

Computer → Information → NLG → Customer Understands
```

### 🔹 **3.6 Evolution of NLP: Rules to Modern Language Models**

**Definition**  
NLP has developed through rule-based methods, statistical learning, neural networks, and pretrained language models. These approaches coexist.

**Example**  
| Approach | Business example | Strength | Limitation |
|---|---|---|---|
| Rules and dictionaries | Route messages containing “refund” | Transparent and controllable | Misses unfamiliar wording |
| Statistical machine learning | Learn complaint categories from labeled reviews | Learns recurring associations | Needs suitable labels and features |
| Neural and contextual models | Use surrounding text to distinguish meanings | Captures more context | Needs evaluation and computing resources |
| Generative language models | Draft a summary of verified complaints | Flexible text generation | May invent or omit details |

**🚨 Problem**  
A business may assume the newest model is automatically the best choice.

**✅ Solution**  
Choose the least complex approach that meets accuracy, cost, latency, privacy, and maintenance requirements. Compare it against a simple baseline.

### 🔹 **3.7 Text Mining vs NLP vs Text Analytics**

**Definition**  
NLP studies computational language processing. Text mining discovers useful patterns across text collections. Text analytics applies measurement and interpretation to text; in practice, the terms overlap.

**Example**  
| Term | Typical question | Example output |
|---|---|---|
| NLP | What does this message communicate? | Refund request with negative sentiment |
| Text mining | Which issues recur across reviews? | Delivery and refund themes |
| Text analytics | How do these issues vary? | Complaint share by channel and period |

**🚨 Problem**  
Treating these as completely separate technologies creates an inaccurate picture of the workflow.

**✅ Solution**  
Use NLP techniques within text-mining and text-analytics projects. Define the required output before selecting the method.

### 🔹 **3.8 AI, Machine Learning, Deep Learning, and Generative AI**

**Definition**  
AI is the broad field. Machine learning learns from data; deep learning uses multilayer neural networks. NLP is a task domain, and generative AI refers to systems that produce content.

**Example**  
A complaint-routing model is an NLP application that may use machine learning. A transformer that drafts support replies combines deep learning, NLP, and generative AI.

**🚨 Problem**  
A simple hierarchy can wrongly imply that NLP and machine learning do not overlap.

**✅ Solution**  
Distinguish the task from the method: NLP can use rules or learned models; generative AI can operate on text, images, or other data.

### 🔹 **3.9 Language Models and Large Language Models**

#### **Definition**
**Generative AI** produces new content such as text, images, audio, or code.

A **Large Language Model (LLM)** is trained on very large amounts of text and can perform many language-related tasks.

#### **Business Uses**
- Summarizing complaints
- Extracting information from reports
- Answering questions about documents
- Drafting customer responses
- Organizing survey feedback
- Identifying themes
- Supporting knowledge search

#### 🚨 **Business Problem**
Managers may need to understand thousands of customer comments quickly.

#### ✅ **Solution**
Modern NLP systems can help summarize and organize the information.

#### ⚠️ **Important Considerations**
- Accuracy
- Privacy
- Bias
- Data security
- Verification

> **Verification:** A fluent summary is not evidence of factual accuracy. Check it against the source comments and retain links to supporting records. Predicting plausible text is different from confirming a business event.

### 🔹 **3.10 NLP Tasks in Business**

**Definition**  
A task is the specific language-processing output needed by a business process.

**Example**  
| Task | Input | Useful output |
|---|---|---|
| Intent detection | “Please return my payment.” | Refund request |
| Entity extraction | “Order 42 reached Lahore.” | Order reference and location |
| Sentiment analysis | “Support was not helpful.” | Negative opinion about support |
| Topic discovery | Many unlabeled complaints | Recurring themes for review |
| Summarization | A set of verified tickets | Concise summary with source references |

**🚨 Problem**  
One overall sentiment score cannot answer every business question.

**✅ Solution**  
Select tasks that produce the evidence needed for the decision. For urgent or sensitive actions, use validated systems and appropriate human review.

## ⚙️ **4. Business Text Mining Workflow**

### 🔹 **4.1 From Data to Information, Insight, and Action**

Consider:

> “The food was excellent, but it arrived 45 minutes late and was almost cold.”

```text
Food Quality   → Positive
Delivery Time  → Negative
Temperature    → Negative
```

#### **Data**
Raw customer statement.

#### **Information**
A delivery delay is present.

#### **Insight**
Delivery delays may be contributing to poor satisfaction.

#### **Action**
Investigate delivery operations in affected areas.

**Process:** Customer Text → Data → Text Processing → Patterns → Information → Business Insight → Decision → Action.

#### 💡 **Key Principle**
The purpose of business text mining is not merely to analyze words.  
**The purpose is to improve decisions.**

### 🔹 **4.2 The End-to-End Workflow**

#### **Definition**
A **text mining workflow** is a structured sequence of steps used to transform raw textual information into useful business insight.

A typical workflow includes:

1. Define the business problem
2. Collect relevant text
3. Clean and prepare the text
4. Represent the text in a machine-processable form
5. Analyze patterns, topics, or sentiment
6. Interpret the results
7. Make a business decision

#### **Example**
A food-delivery company wants to understand why customers are leaving poor ratings.

**Process:** Business Problem → Collect Customer Reviews → Prepare Text → Identify Complaint Topics → Measure Negative Feedback → Interpret Results → Improve Operations.

#### 🚨 **Problem**
Without a clear workflow, analysts may jump directly into coding or modeling before understanding the business question.

#### ✅ **Solution**
Start with the business problem, then select data and techniques that directly support that problem.

#### 💡 **Business Lesson**
The best technical model is not useful if it does not answer a meaningful business question.

### 🔹 **4.3 Define the Business Problem**

#### **Definition**
The first step is to identify the decision, issue, or opportunity the organization wants to understand.

#### **Example**
A retailer notices an increase in product returns.

A weak question:

> “Can we analyze customer text?”

A better question:

> “What are the main reasons customers return our products, and which reasons are increasing?”

#### 🚨 **Problem**
Vague objectives lead to irrelevant analysis.

#### ✅ **Solution**
Turn the business problem into a specific analytical question.

#### **Useful Business Questions**
- What are customers complaining about?
- Which issues are increasing?
- What features receive positive feedback?
- Which service area causes the most dissatisfaction?
- How do customer concerns differ across regions?
- What topics are associated with refunds or churn?

### 🔹 **4.4 Collect Relevant Text**

#### **Definition**
Text collection means acquiring textual information from sources that are relevant to the business question.

#### **Example**
For a customer-experience problem, useful sources may include:

- Product reviews
- Complaint tickets
- Survey comments
- Chat transcripts
- Social media posts
- App reviews

#### 🚨 **Problem**
Collecting too much irrelevant text creates noise.

#### ✅ **Solution**
Select text sources that directly relate to the business objective.

> **Business objective first. Data source second.**

### 🔹 **4.5 Prepare Text: Conceptual Overview**

#### **Definition**
Raw text often requires preparation before analysis.

This may include:

- Lowercasing
- Removing unnecessary symbols
- Removing URLs
- Tokenization
- Stop-word handling
- Lemmatization
- Removing duplicates
- Handling missing values

#### **Example**
**Process:** Raw: → "AMAZING product!!! 😍 Visit https://shop.com" → Prepared: → "amazing product".

#### 🚨 **Problem**
Noisy text can distort the results.

#### ✅ **Solution**
Use a preprocessing pipeline appropriate for the business objective.

> Detailed preprocessing is developed in **Week 11**.

### 🔹 **4.6 Represent Text: Conceptual Overview**

#### **Definition**
Computers need text to be converted into a numerical or structured representation before many analytical techniques can be applied.

#### **Examples**
Text can be represented using:

- Word counts
- Term frequencies
- TF-IDF
- Word embeddings
- Sentence embeddings
- Transformer-based representations

#### 🚨 **Problem**
Raw sentences cannot be directly used by many traditional machine-learning algorithms.

#### ✅ **Solution**
Convert language into numerical features while preserving useful information.

**Process:** Text → Representation → Numbers / Vectors → Machine Learning.

> Advanced representation methods are developed later in the course.

### 🔹 **4.7 Analyze Patterns and Meaning**

#### **Definition**
This step applies analytical methods to discover useful patterns in the prepared text.

#### **Examples**
Possible tasks include:

- Sentiment analysis
- Topic discovery
- Complaint categorization
- Text categorization
- Keyword extraction
- Similarity analysis
- Trend detection

#### 🚨 **Problem**
A company may have millions of comments but no clear summary.

#### ✅ **Solution**
Use suitable text-mining techniques to organize and quantify recurring patterns.

### 🔹 **4.8 Interpret and Validate Results**

#### **Definition**
Interpretation means translating technical outputs into business meaning.

#### **Example**
Technical result:

```text
Topic 1:
delivery, late, delay, rider, order
```

Business interpretation:

> **Delivery Delay Complaints**

#### 🚨 **Problem**
Technical outputs may be difficult for managers to understand.

#### ✅ **Solution**
Translate technical findings into clear business language.

> **Model output is not the final result. Business interpretation is.**

Review actual sentences before naming a topic. A mention of “late” may occur in “not late.” Define the denominator and compare like-for-like periods and channels.

### 🔹 **4.9 Take Action and Monitor Outcomes**

#### **Definition**
The final step is to use the insight to support a decision or operational improvement.

#### **Example**
Finding:

> Delivery-delay complaints increased in the observed sample in one region.

Possible actions:

- Review delivery partners
- Increase rider capacity
- Adjust delivery estimates
- Improve order-routing logic
- Monitor improvement after intervention

#### 🚨 **Problem**
Text analysis that does not influence decisions produces little business value.

#### ✅ **Solution**
Connect every major finding with a possible business response.

### 🔹 **4.10 Sampling, Bias, Evaluation, and Reporting**

**Definition**  
A reliable text-mining report states who supplied the text, which records were included, how outputs were checked, and what the results can support.

**Example**  
In an illustrative sample of 100 unique reviews, 25 mention delivery and 10 describe a delivery problem. Delivery mention share is 25%; delivery complaint share is 10%. Neither is automatically the delivery-failure rate among all orders.

**🚨 Problem**  
Duplicates, fake reviews, missing languages, channel changes, and model mistakes can produce misleading trends.

**✅ Solution**  
Audit representative examples; compare model labels with reviewed labels; report sample coverage and the exact denominator. Separate frequency, sentiment, severity, and operational impact.

**Report checklist:** State the business question, source and period, inclusion rules, method, validated findings, limitations, recommended action, owner, and follow-up measure. A bar chart can compare topics; a line chart can show a comparable trend. Word size alone is insufficient evidence of importance.

### **Step 2 — Extract Review Text and Metadata from HTML**

`select()` locates containers, `select_one()` locates a field, and `get_text()` extracts its text. The parser retains the review ID so repeated collection can be detected later. A missing expected field is reported rather than silently skipped.

````python
soup = BeautifulSoup(html_page, "html.parser")  # Parse the embedded webpage.
web_rows = []  # Accumulate extracted review records.
for article in soup.select("article[data-review-id]"):  # Restrict extraction to review containers.
    text_node = article.select_one("p[data-text]")  # Locate the review-text field.
    if text_node is None:  # Detect an unexpected page structure.
        raise ValueError("A review container has no text field.")  # Stop rather than silently lose a record.
    web_rows.append({  # Map the HTML fields to a common schema.
        "source": "web", "source_record_id": article["data-review-id"],  # Retain provenance and source identity.
        "business_area": article["data-area"], "language": article["data-language"],  # Keep interpretation metadata.
        "rating": int(article["data-rating"]), "raw_text": text_node.get_text(),  # Preserve text content without NLP cleaning.
    })  # Finish one extracted record.
web_reviews = pd.DataFrame(web_rows)  # Convert the collected records into a table.
display(web_reviews.head(3))  # Inspect a small sample of the extraction.
````

### **Step 3 — Parse JSON and Build One Consistent Dataset**

JSON is serialized text; `json.loads()` converts it into Python objects. The same field names across sources allow a combined table. Each collection row receives an `ingestion_id`, while `source + source_record_id` identifies the original review.

````python
payload = json.loads(api_json)  # Deserialize the embedded JSON response.
if not isinstance(payload.get("reviews"), list):  # Check the expected response structure.
    raise ValueError("The response must contain a reviews list.")  # Make a schema problem visible.
api_reviews = pd.DataFrame(payload["reviews"])  # Convert the records to a table.
fields = ["source", "source_record_id", "business_area", "language", "rating", "raw_text"]  # Define common collection fields.
reviews = pd.concat([web_reviews[fields], api_reviews[fields]], ignore_index=True)  # Combine both sources.
reviews.insert(0, "ingestion_id", [f"I{i:03d}" for i in range(1, len(reviews) + 1)])  # Identify each collected row.
reviews["review_key"] = reviews["source"] + ":" + reviews["source_record_id"]  # Build a stable source-scoped review key.
display(reviews[["ingestion_id", "review_key", "business_area", "raw_text"]].head(6))  # Preview the common schema.
````

### **Step 4 — Inspect Quality Without Changing the Evidence**

A missing comment is not a neutral opinion. The same review fetched twice is not two independent customer opinions. Flag these issues now; Week 11 will decide how to handle them.

````python
missing_text = reviews["raw_text"].isna()  # Identify absent text values.
blank_text = reviews["raw_text"].fillna("").str.strip().eq("") & ~missing_text  # Identify whitespace-only text separately.
duplicate_key = reviews.duplicated("review_key", keep="first")  # Find repeated collection of a source record.
quality = pd.DataFrame({"Check": ["Collected rows", "Unique review keys", "Missing text", "Blank text", "Repeated review key"],  # Name each measure.
                        "Count": [len(reviews), reviews["review_key"].nunique(), int(missing_text.sum()), int(blank_text.sum()), int(duplicate_key.sum())]})  # Compute the measures.
display(quality)  # Display the collection audit.
assert len(reviews) == 16 and missing_text.sum() == 1 and blank_text.sum() == 1  # Verify the fixture was extracted correctly.
assert duplicate_key.sum() == 1  # Verify repeated collection remains visible for Week 11.
````

### **Step 7 — Connect the Demonstration to Real HTTP Collection**

The helpers below show the request stage that precedes the HTML and JSON parsing. They are defined for practice and are **not invoked by Run All**. Supply only a URL you are permitted to collect; no external endpoint or credential is assumed.

**Before use:** Check access rules, pagination, field definitions, and rate limits. `raise_for_status()` catches unsuccessful HTTP status codes; a successful status still requires content and schema checks. A JSON response may contain only one page.

````python
import requests  # Make HTTP requests when a permitted source is available.
def fetch_html(url):  # Fetch one permitted HTML page for the same parsing workflow.
    response = requests.get(url, timeout=15)  # Bound network waiting rather than waiting indefinitely.
    response.raise_for_status()  # Report unsuccessful HTTP status codes.
    return response.text  # Return HTML to BeautifulSoup for source-specific extraction.
def fetch_json(url, params=None):  # Fetch one permitted API response.
    response = requests.get(url, params=params, timeout=15)  # Send documented query parameters with a timeout.
    response.raise_for_status()  # Check HTTP status before interpreting the response.
    return response.json()  # Decode JSON; invalid JSON raises a visible error.
print("HTTP helpers defined. The executed collection used the embedded HTML and JSON fixtures.")  # Distinguish tested extraction from optional network use.
````
