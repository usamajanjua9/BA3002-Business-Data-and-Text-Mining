## **🟢 What is Data?**

Data is the **raw collection of facts, figures, observations, transactions, or records** that a business gathers.

By itself, data may not immediately tell us anything useful.

Think of data as the **raw material** of Business Analytics.

### 🔹 **Examples of Business Data**

A supermarket may collect:

* Customer ID
* Product purchased
* Quantity
* Price
* Date and time
* Payment method
* Store location
* Customer rating

For example:

| **Customer** | **Product** | **Quantity** | **Amount (Rs.)** |
| ------------ | ----------- | -----------: | ---------------: |
| C001         | Coffee      |            2 |            1,200 |
| C002         | Milk        |            4 |            1,000 |
| C003         | Coffee      |            1 |              600 |
| C004         | Tea         |            3 |            1,500 |

These numbers and labels are **data**.

At this stage, we have not yet interpreted them.

## **🟢 What is Information?**

Raw data becomes **information** when it is:

* Organized
* Processed
* Summarized
* Given context
* Made meaningful

---

### 🔹 Example

Suppose we have the following raw sales data:

| Product | Units Sold |
| ------- | ---------: |
| Coffee  |         50 |
| Tea     |         30 |
| Milk    |         20 |

This is **data**.

After analyzing it, we conclude:

> **Coffee is the highest-selling product.**

That statement is **information**.

---

### 📍 **Simple Definition**

> **Information = Processed data that has meaning.**

---

## **🔄 Data vs Information**

| **Data**                      | **Information**                                |
| ----------------------------- | ---------------------------------------------- |
| Raw facts                     | Processed facts                                |

| Customer rating = 1           | Customer satisfaction is declining             |

## **🟢 What is Knowledge?**

Knowledge is created when we **interpret information and understand what it means for the business**.

It answers questions such as:

### 🤔 Why is this happening?

### 🔍 What pattern exists?

### 🎯 What should we learn from it?

### 💡 How can we use this information?

---

### 🔹 Example

Suppose our analysis shows:

> **Coffee sales are 40% higher on weekday mornings.**

That is useful **information**.

After studying customer behaviour, we discover:

> **Office workers purchase coffee mainly between 8:00 AM and 10:00 AM.**

Now we have **knowledge** about customer behaviour.

The business can use this knowledge to:

* Increase coffee inventory in the morning
* Offer breakfast bundles
* Target office workers
* Schedule additional staff

---

### 📍 **Simple Definition**

> **Knowledge = Understanding obtained by interpreting information and identifying meaningful patterns.**

---

## **🟢 Data → Information → Knowledge**

Let's understand the difference using one simple example.

### 🛒 Raw Data

A retailer records:

`Milk — Bread — Eggs`

`Milk — Bread`

`Milk — Eggs`

`Bread — Milk`

These transaction records are **DATA**.

⬇️

### 📊 Information

After analyzing thousands of transactions:

> **60% of customers who buy bread also buy milk.**

This is **INFORMATION**.

⬇️

### 🧠 Knowledge

The business realizes:

> **Bread and milk have a strong purchasing relationship.**

This is **KNOWLEDGE**.

⬇️

### 🎯 Decision

The manager decides to:

> **Place bread and milk strategically or offer a bundle promotion.**

This is a **BUSINESS DECISION**.

* Rows
* Columns
* Tables

---

## **🔹 Example**

| Customer ID | Age | City      | Spending |
| ----------- | --: | --------- | -------: |
| C001        |  23 | Islamabad |   35,000 |
| C002        |  31 | Lahore    |   58,000 |
| C003        |  28 | Karachi   |   42,000 |

Every row follows the same structure.

---

### 📍 **Simple Definition**

> **Structured Data = Data stored in a predefined and organized format.**

---

## **Examples of Structured Business Data**

* Sales databases
* Customer records
* Banking transactions
* Inventory tables
* Financial statements
* Employee databases
* Excel spreadsheets
* SQL databases

# **🟢 Why is Structured Data Easy to Analyze?**

* Which column represents age
* Which column represents price
* Which column represents customer ID

It can easily:

* Sort
* Filter
* Calculate
* Aggregate
* Build models

# **🟢 Semi-Structured Data**

Semi-Structured Data does not follow a traditional row-and-column format, but it still contains **labels, tags, keys, or metadata**.

Common examples include:

* JSON
* XML
* Emails
* API responses
* Web logs

---

## **🔹 JSON Example**

```json
{
  "customer_id": "C001",
  "name": "Ali",
  "city": "Islamabad",
  "spending": 35000
}
```

This is not a traditional table.

But it still has a recognizable structure:

`customer_id`

`name`

`city`

`spending`

---

### 📍 **Simple Definition**

> **Semi-Structured Data = Data that has some organization, but does not follow a strict tabular structure.**

---

# **🟢 Business Examples of Semi-Structured Data**

### 🔌 API Response

A company requests currency information from an API.

### 📧 Email

An email has fields such as:

* Sender
* Receiver

* Date

### 🌐 Website HTML

A webpage may contain:

```html

<p>Price: Rs. 150,000</p>
```

It has tags and structure, but it is not stored as a traditional business table.

## **🟢 What is Data Mining?**

Data Mining is the process of discovering **useful patterns, relationships, trends, and hidden knowledge** from large amounts of data.

The aim is not just to look at data.

The aim is to find something **useful, meaningful, and actionable**.

---

### 📍 **Simple Definition**

> **Data Mining = Discovering useful patterns and knowledge hidden inside data.**

---

## **🔹 Think of It Like Traditional Mining**

In traditional mining:

**⛏️ Rock → Search → Valuable Material**

In Data Mining:

**💾 Data → Analysis → Valuable Knowledge**

The valuable material may be:

* A hidden customer segment
* A fraud pattern
* A purchasing relationship
* A churn indicator
* A suspicious transaction
* A sales trend
* A recommendation opportunity

---

## **🧠 Example**

Suppose a supermarket has:

### **2 million transactions**

At first, these are simply records.

But after mining the data, we may discover:

> Customers who buy baby diapers frequently also purchase baby wipes.

This is no longer just raw data.

This is a **pattern**.

The business can now:

* Bundle the products
* Offer discounts
* Improve shelf placement
* Recommend related items

Businesses collect huge amounts of data every day.

The problem is:

> **Most of that data is useless unless we can extract something meaningful from it.**

Businesses want to know:

* Who are our best customers?

* What products do they prefer?
* Which customers respond to promotions?

A telecom company may discover:

> Customers with frequent complaints and high monthly bills are more likely to cancel.

## **🔹 2. To Increase Sales 🛒**

Data Mining can identify:

* Products purchased together
* High-value customers
* Cross-selling opportunities
* Seasonal buying patterns

### Example

An online retailer may discover:

> Customers buying laptops often purchase wireless mice within seven days.

## **🔹 3. To Detect Fraud 🕵️**

Banks and financial companies use Data Mining to identify unusual behavior.

Examples:

* Sudden high-value transactions

* Multiple rapid transactions
* Unexpected spending patterns

# **🟢 Data Mining vs Data Analysis**

These two terms are related, but they are not exactly the same.

---

## **🔹 Data Analysis**

Data Analysis focuses on examining data to answer specific questions.

Example:

> What were our sales last month?

> Which branch performed best?

> Why did revenue fall?

---

## **🔹 Data Mining**

Data Mining focuses more on discovering patterns that may not already be known.

Example:

> Which customers behave similarly?

> Which products are frequently bought together?

> Which customers are likely to leave?

---

## **📊 Comparison**

| **Data Analysis**                 | **Data Mining**                               |
| --------------------------------- | --------------------------------------------- |
| Answers known questions           | Discovers hidden patterns                     |
| Often descriptive                 | Often exploratory/predictive                  |
| Uses statistics and visualization | Uses algorithms and pattern discovery         |
| Example: Why did sales fall?      | Example: Which customers are likely to churn? |

---

# **🟢 Data Mining vs Machine Learning**

These areas overlap heavily.

---

## **🔹 Machine Learning**

Machine Learning focuses on building systems that learn patterns from data and make predictions or decisions.

Example:

> Predict whether a customer will churn.

---

## **🔹 Data Mining**

Data Mining has a broader objective:

> Discover useful knowledge from data.

It may use Machine Learning algorithms to do this.

---

## **📍 Simple Way to Remember**

> **Machine Learning learns from data.**

> **Data Mining uses data to discover useful knowledge.**

---

## **📊 Comparison**

| **Data Mining**                 | **Machine Learning**                    |
| ------------------------------- | --------------------------------------- |
| Focuses on knowledge discovery  | Focuses on learning predictive patterns |
| Often business-oriented         | Often model-oriented                    |
| May use ML algorithms           | Includes many prediction algorithms     |
| Example: Find customer segments | Example: Predict customer churn         |

---

## **🟢 What is Descriptive Data Mining?**

Descriptive Data Mining focuses on discovering **patterns, relationships, groups, and trends** that already exist in data.

It does not necessarily try to predict the future.

Instead, it asks:

> **What interesting structure or pattern exists inside our data?**

---

### 📍 **Simple Definition**

> **Descriptive Data Mining = Discovering meaningful patterns in existing data.**

---

# **🟢 1. Pattern Discovery**

Pattern discovery means finding **repeated or meaningful behavior** hidden inside large datasets.

Businesses may discover:

* Frequently purchased products
* Similar customer behavior
* Common complaint themes
* Repeated transaction patterns
* Seasonal changes in demand

---

## **🔹 Example**

A retailer analyzes thousands of orders and discovers:

> Customers who buy smartphones often purchase screen protectors within the same order.

This is a **pattern**.

The company can use it for:

* Bundling
* Recommendations
* Cross-selling

### 🛍️ **Regular Customers**

### 💤 **Low-Engagement Customers**

### 💡 Business Use

These groups can help businesses create:

* Personalized communication

Association discovery finds items or events that occur **together frequently**.

## **🟢 What is Predictive Data Mining?**

Predictive Data Mining uses **historical data** to estimate what is likely to happen in the future.

It asks questions such as:

* Will this customer leave?
* Will this borrower default?
* How much will we sell next month?
* Is this transaction likely to be fraudulent?

---

### 📍 **Simple Definition**

> **Predictive Data Mining = Learning from past data to predict future outcomes.**

---

# **🟢 Prediction vs Classification**

These terms are related but not identical.

---

## **🔹 Classification**

Classification predicts a **category or class**.

Examples:

* Churn: Yes / No
* Fraud: Fraud / Not Fraud
* Loan: Approve / Reject
* Email: Spam / Not Spam

---

## **🔹 Prediction**

Prediction may estimate a **future numeric value**.

Examples:

* Next month's sales
* Product demand
* Customer lifetime value
* Revenue forecast

---

## **📊 Quick Comparison**

| **Task**       | **Output**    | **Example**              |
| -------------- | ------------- | ------------------------ |
| Classification | Category      | Customer will churn: Yes |
| Prediction     | Numeric value | Expected sales: Rs. 2.4M |

---

# **🟢 Using Historical Data to Predict the Future**

Predictive Data Mining learns from past observations.

For example:

A telecom company may have records of customers who:

* Stayed
* Left
* Complained
* Paid late
* Changed packages

The model learns the relationship between these factors and churn.

Then it applies those learned patterns to **new customers**.

---

## **🔄 Simple Flow**

**Historical Data**

⬇️

**Learn Patterns**

⬇️

**Build Model**

⬇️

**New Customer**

⬇️

**Prediction**

````python
# ============================================================
# BA3002 - BUSINESS DATA AND TEXT MINING
# WEEK 1 PRACTICAL
# Topic: Understanding a Simple Business Dataset
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 2. CREATE A SIMPLE CUSTOMER DATASET
# ------------------------------------------------------------

data = {
    "Customer_ID": ["C001", "C002", "C003", "C004", "C005", "C006"],

    "Age": [23, 31, 27, 45, 36, 29],

    "City": [
        "Islamabad",
        "Lahore",
        "Islamabad",
        "Karachi",
        "Lahore",
        "Karachi"
    ],

    "Monthly_Spending": [
        18000,
        9500,
        26000,
        7000,
        32000,
        12500
    ],

    "Orders": [6, 2, 8, 1, 10, 3],

    "Complaints": [0, 4, 1, 5, 0, 3],

    "Review": [
        "Excellent service and fast delivery",
        "Delivery was late and support was poor",
        "Good products and reasonable prices",
        "Very disappointed with the service",
        "Great experience, will order again",
        "Product was fine but delivery was slow"
    ],

    "Churn": [
        "No",
        "Yes",
        "No",
        "Yes",
        "No",
        "Yes"
    ]
}


# ------------------------------------------------------------
# 3. CONVERT THE DATA INTO A PANDAS DATAFRAME
# ------------------------------------------------------------

df = pd.DataFrame(data)


# ------------------------------------------------------------
# 4. DISPLAY THE COMPLETE DATASET
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("1. CUSTOMER DATASET")
print("=" * 70)

print(df)


# ------------------------------------------------------------
# 5. CHECK DATASET SIZE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("2. DATASET SIZE")
print("=" * 70)

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# ------------------------------------------------------------
# 6. DISPLAY COLUMN NAMES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("3. VARIABLES / COLUMN NAMES")
print("=" * 70)

print(df.columns.tolist())


# ------------------------------------------------------------
# 7. CHECK DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("4. DATA TYPES")
print("=" * 70)

print(df.dtypes)


# ------------------------------------------------------------
# 8. DISPLAY DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("5. DATASET INFORMATION")
print("=" * 70)

df.info()


# ------------------------------------------------------------
# 9. DISPLAY BASIC NUMERICAL STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("6. BASIC STATISTICS")
print("=" * 70)

print(df.describe())


# ------------------------------------------------------------
# 10. CALCULATE AVERAGE MONTHLY SPENDING
# ------------------------------------------------------------

average_spending = df["Monthly_Spending"].mean()

print("\n" + "=" * 70)
print("7. AVERAGE MONTHLY SPENDING")
print("=" * 70)

print(f"Average Monthly Spending: Rs. {average_spending:,.2f}")


# ------------------------------------------------------------
# 11. CALCULATE AVERAGE NUMBER OF ORDERS
# ------------------------------------------------------------

average_orders = df["Orders"].mean()

print("\n" + "=" * 70)
print("8. AVERAGE NUMBER OF ORDERS")
print("=" * 70)

print(f"Average Orders per Customer: {average_orders:.2f}")


# ------------------------------------------------------------
# 12. COUNT CUSTOMERS BY CITY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("9. CUSTOMERS BY CITY")
print("=" * 70)

print(df["City"].value_counts())


# ------------------------------------------------------------
# 13. COUNT CHURNED AND NON-CHURNED CUSTOMERS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("10. CHURN STATUS")
print("=" * 70)

print(df["Churn"].value_counts())


# ------------------------------------------------------------
# 14. CALCULATE CHURN PERCENTAGE
# ------------------------------------------------------------

churn_percentage = (
    df["Churn"]
    .value_counts(normalize=True)
    .mul(100)
)

print("\n" + "=" * 70)
print("11. CHURN PERCENTAGE")
print("=" * 70)

print(churn_percentage.round(2))


# ------------------------------------------------------------
# 15. COMPARE AVERAGE COMPLAINTS BY CHURN STATUS
# ------------------------------------------------------------

average_complaints_by_churn = (
    df.groupby("Churn")["Complaints"]
    .mean()
)

print("\n" + "=" * 70)
print("12. AVERAGE COMPLAINTS BY CHURN STATUS")
print("=" * 70)

print(average_complaints_by_churn)


# ------------------------------------------------------------
# 16. COMPARE AVERAGE SPENDING BY CHURN STATUS
# ------------------------------------------------------------

average_spending_by_churn = (
    df.groupby("Churn")["Monthly_Spending"]
    .mean()
)

print("\n" + "=" * 70)
print("13. AVERAGE SPENDING BY CHURN STATUS")
print("=" * 70)

print(average_spending_by_churn)


# ------------------------------------------------------------
# 17. COMPARE AVERAGE ORDERS BY CHURN STATUS
# ------------------------------------------------------------

average_orders_by_churn = (
    df.groupby("Churn")["Orders"]
    .mean()
)

print("\n" + "=" * 70)
print("14. AVERAGE ORDERS BY CHURN STATUS")
print("=" * 70)

print(average_orders_by_churn)


# ------------------------------------------------------------
# 18. FIND HIGH-COMPLAINT CUSTOMERS
# ------------------------------------------------------------

high_complaints = df[
    df["Complaints"] >= 3
]

print("\n" + "=" * 70)
print("15. HIGH-COMPLAINT CUSTOMERS")
print("=" * 70)

print(
    high_complaints[
        [
            "Customer_ID",
            "Complaints",
            "Monthly_Spending",
            "Orders",
            "Churn"
        ]
    ]
)


# ------------------------------------------------------------
# 19. FIND CUSTOMERS WHO HAVE CHURNED
# ------------------------------------------------------------

churned_customers = df[
    df["Churn"] == "Yes"
]

print("\n" + "=" * 70)
print("16. CHURNED CUSTOMERS")
print("=" * 70)

print(
    churned_customers[
        [
            "Customer_ID",
            "City",
            "Monthly_Spending",
            "Complaints",
            "Churn"
        ]
    ]
)


# ------------------------------------------------------------
# 20. FIND REVIEWS THAT MENTION "DELIVERY"
# ------------------------------------------------------------

delivery_reviews = df[
    df["Review"].str.contains(
        "delivery",
        case=False,
        na=False
    )
]

print("\n" + "=" * 70)
print("17. REVIEWS MENTIONING 'DELIVERY'")
print("=" * 70)

print(
    delivery_reviews[
        [
            "Customer_ID",
            "Review",
            "Churn"
        ]
    ]
)


# ------------------------------------------------------------
# 21. CREATE A NEW VARIABLE: COMPLAINT LEVEL
# ------------------------------------------------------------

df["Complaint_Level"] = df["Complaints"].apply(
    lambda x: "High" if x >= 3 else "Low"
)

print("\n" + "=" * 70)
print("18. COMPLAINT LEVEL")
print("=" * 70)

print(
    df[
        [
            "Customer_ID",
            "Complaints",
            "Complaint_Level"
        ]
    ]
)


# ------------------------------------------------------------
# 22. DISPLAY A SIMPLE BUSINESS SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("19. BUSINESS SUMMARY")
print("=" * 70)

print("Total Customers:", len(df))

print(
    f"Average Monthly Spending: "
    f"Rs. {df['Monthly_Spending'].mean():,.2f}"
)

print(
    f"Average Orders per Customer: "
    f"{df['Orders'].mean():.2f}"
)

print(
    f"Average Complaints per Customer: "
    f"{df['Complaints'].mean():.2f}"
)

print(
    "Number of Churned Customers:",
    (df["Churn"] == "Yes").sum()
)


# ============================================================
# VISUALIZATION SECTION
# ============================================================


# ------------------------------------------------------------
# 23. BAR CHART - MONTHLY SPENDING BY CUSTOMER
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    df["Customer_ID"],
    df["Monthly_Spending"]
)

plt.title("Monthly Spending by Customer")
plt.xlabel("Customer ID")
plt.ylabel("Monthly Spending (Rs.)")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 24. BAR CHART - COMPLAINTS BY CUSTOMER
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    df["Customer_ID"],
    df["Complaints"]
)

plt.title("Number of Complaints by Customer")
plt.xlabel("Customer ID")
plt.ylabel("Number of Complaints")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 25. BAR CHART - AVERAGE COMPLAINTS BY CHURN STATUS
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

average_complaints_by_churn.plot(
    kind="bar"
)

plt.title("Average Complaints by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Average Number of Complaints")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 26. BAR CHART - AVERAGE SPENDING BY CHURN STATUS
# ------------------------------------------------------------

plt.figure(figsize=(7, 5))

average_spending_by_churn.plot(
    kind="bar"
)

plt.title("Average Monthly Spending by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Average Monthly Spending (Rs.)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 27. FINAL BUSINESS INTERPRETATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("20. FINAL BUSINESS INTERPRETATION")
print("=" * 70)

print(
    """
Possible observations from this small dataset:

1. Customers with more complaints appear more likely to churn.

2. Customers who remain active generally have more orders.

3. Non-churned customers appear to have higher monthly spending.

4. Several negative customer reviews mention delivery and service issues.

5. Complaint level may be an important variable for understanding churn.

Important:
These are only observations from a very small dataset.
They do not prove that complaints directly cause churn.

The purpose of this practical is to learn how to move from:

DATA -> ANALYSIS -> PATTERN -> INSIGHT -> BUSINESS QUESTION
"""
)
````

# **📘 Week 1-2 — Main Definitions for Revision**

---

## **1. Data**

**Data** is a collection of raw facts, figures, observations, transactions, or records that have not yet been interpreted or given meaning.

**Example:**
`Customer Age = 25`, `Sales = Rs. 50,000`, `Rating = 4`

---

## **2. Information**

**Information** is data that has been organized, processed, summarized, or placed in context so that it becomes meaningful.

**Example:**
“Sales increased by 15% this month.”

---

## **3. Knowledge**

**Knowledge** is the understanding gained by interpreting information, identifying patterns, and recognizing their significance for a business.

**Example:**
“Weekend promotions are responsible for the increase in sales.”

---

## **4. Business Data**

**Business Data** is any data generated, collected, stored, or used by an organization to understand its customers, operations, products, markets, finances, or performance.

---

## **5. Internal Data**

**Internal Data** is data generated from within an organization through its own systems, processes, customers, employees, and operations.

**Examples:**
Sales records, CRM data, transaction history, inventory records.

---

## **6. External Data**

**External Data** is data obtained from sources outside an organization to understand markets, competitors, customers, economic conditions, or public opinion.

**Examples:**
Websites, social media, government statistics, market reports.

---

## **7. Structured Data**

**Structured Data** is data organized according to a predefined format or schema, usually in rows and columns.

**Examples:**
Excel sheets, SQL tables, transaction databases.

---

## **8. Semi-Structured Data**

**Semi-Structured Data** is data that does not follow a strict tabular structure but contains some organization through tags, keys, labels, or metadata.

**Examples:**
JSON, XML, HTML, API responses.

---

## **9. Unstructured Data**

**Unstructured Data** is data that does not follow a predefined organizational structure and is usually difficult to analyze directly using traditional tabular methods.

**Examples:**
Customer reviews, emails, images, videos, social-media comments.

---

## **10. Data Mining**

**Data Mining** is the process of discovering useful patterns, relationships, trends, and knowledge from large datasets using analytical and computational techniques.

---

## **11. Pattern Discovery**

**Pattern Discovery** is the process of identifying repeated, meaningful, or previously unknown relationships and behaviors within data.

**Example:**
Customers who buy laptops frequently also buy wireless mice.

---

## **12. Knowledge Discovery in Databases (KDD)**

**KDD** is the complete process of converting raw data into useful knowledge through data selection, preparation, mining, evaluation, and interpretation.

---

## **13. Business Analytics**

**Business Analytics** is the use of data, statistical methods, analytical techniques, and computational tools to understand business performance and support better decisions.

---

## **14. Descriptive Analytics**

**Descriptive Analytics** analyzes historical data to summarize and explain **what has already happened**.

**Main Question:**

> What happened?

---

## **15. Diagnostic Analytics**

**Diagnostic Analytics** examines data to determine the reasons and factors behind an observed event or outcome.

**Main Question:**

> Why did it happen?

---

## **16. Predictive Analytics**

**Predictive Analytics** uses historical data and analytical models to estimate future events or outcomes.

**Main Question:**

> What is likely to happen?

---

## **17. Prescriptive Analytics**

**Prescriptive Analytics** uses data, predictions, business objectives, and constraints to recommend actions that should be taken.

**Main Question:**

> What should we do?

---

## **18. Descriptive Data Mining**

**Descriptive Data Mining** focuses on discovering existing patterns, relationships, groups, associations, and trends within data.

---

## **19. Predictive Data Mining**

**Predictive Data Mining** uses historical data and learned patterns to estimate future outcomes or classify new observations.

---

## **20. Classification**

**Classification** is a predictive technique used to assign an observation to a predefined category or class.

**Examples:**

* Churn: Yes / No
* Fraud: Fraud / Not Fraud
* Loan: Approve / Reject

---

## **21. Prediction**

**Prediction** is the process of estimating an unknown or future outcome based on patterns learned from historical data.

**Example:**
Predicting next month's sales.

---

## **22. Clustering**

**Clustering** is an unsupervised data mining technique that groups observations based on similarity without using predefined class labels.

**Example:**
Grouping customers into high-value, regular, and low-engagement segments.

---

## **23. Customer Segmentation**

**Customer Segmentation** is the process of dividing customers into groups based on similar characteristics, behaviors, needs, or purchasing patterns.

---

## **24. Association Rule Mining**

**Association Rule Mining** is a data mining technique used to discover relationships between items or events that frequently occur together.

**Example:**
Customers who buy bread often also buy milk.

---

## **25. Market Basket Analysis**

**Market Basket Analysis** is the application of association rule mining to transaction data in order to identify products that are frequently purchased together.

---

## **26. Trend Identification**

**Trend Identification** is the process of detecting consistent increases, decreases, seasonal patterns, or changes in data over time.

---

## **27. Customer Churn**

**Customer Churn** occurs when a customer stops purchasing from, subscribing to, or using the services of a business.

---

## **28. Churn Prediction**

**Churn Prediction** is the use of historical customer data to estimate which customers are likely to leave a business in the future.

---

## **29. Credit Risk**

**Credit Risk** is the possibility that a borrower may fail to repay a loan or financial obligation.

---

## **30. Credit-Risk Prediction**

**Credit-Risk Prediction** is the use of historical financial and customer data to estimate the likelihood that a borrower may default.

---

## **31. Fraud Detection**

**Fraud Detection** is the process of identifying transactions or activities that appear suspicious, abnormal, or potentially fraudulent.

---

## **32. Anomaly**

An **Anomaly** is an observation, transaction, or event that significantly differs from normal or expected behavior.

---

## **33. Anomaly Detection**

**Anomaly Detection** is the process of identifying unusual observations or patterns that differ from normal behavior.

---

## **34. Data Analysis**

**Data Analysis** is the process of examining, transforming, summarizing, and interpreting data to answer specific questions and generate insights.

---

## **35. Machine Learning**

**Machine Learning** is a branch of artificial intelligence in which algorithms learn patterns from data to make predictions, classifications, or decisions without being explicitly programmed for every case.

---

## **36. Business Intelligence (BI)**

**Business Intelligence** refers to technologies and processes used to collect, summarize, visualize, and monitor business data through reports, dashboards, and key performance indicators.

---

## **37. Business Problem**

A **Business Problem** is a challenge, issue, opportunity, or objective that affects organizational performance and requires a decision or solution.

---

## **38. Analytical Problem**

An **Analytical Problem** is a business problem translated into a form that can be addressed using data and analytical techniques.

**Example:**
Business Problem: Customers are leaving.
Analytical Problem: Predict which customers are likely to churn.

---

## **39. Data Collection**

**Data Collection** is the process of gathering relevant data from sources such as files, databases, APIs, websites, surveys, or business systems.

---

## **40. Data Preparation**

**Data Preparation** is the process of cleaning, transforming, integrating, and organizing raw data so that it is suitable for analysis or modeling.

---

## **41. Data Cleaning**

**Data Cleaning** is the process of identifying and correcting problems such as missing values, duplicates, incorrect values, and inconsistent formats.

---

## **42. Data Exploration**

**Data Exploration** is the process of examining a dataset using summaries, statistics, and visualizations to understand its characteristics, patterns, and possible relationships.

---

## **43. Exploratory Data Analysis (EDA)**

**Exploratory Data Analysis** is the systematic examination of data using statistical summaries and visualizations to understand distributions, relationships, trends, and anomalies before modeling.

---

## **44. Variable**

A **Variable** is a characteristic, attribute, or feature recorded for each observation in a dataset.

**Examples:**
Age, City, Spending, Complaints.

---

## **45. Numerical Data**

**Numerical Data** represents quantities expressed as numbers on which mathematical operations can usually be performed.

**Examples:**
Age, income, sales, number of orders.

---

## **46. Categorical Data**

**Categorical Data** represents labels, groups, or categories rather than numerical quantities.

**Examples:**
City, product category, churn status.

---

## **47. Text Data**

**Text Data** consists of words, sentences, documents, reviews, comments, or other natural-language content.

---

## **48. Data Mining Workflow**

A **Data Mining Workflow** is the sequence of steps followed to solve a data-driven business problem.

A simple workflow is:

> **Business Understanding → Data Collection → Data Preparation → Data Exploration → Data Mining → Evaluation → Business Decision**

---

## **49. Evaluation**

**Evaluation** is the process of determining whether a model or discovered pattern is accurate, reliable, meaningful, and useful for the business problem.

---

## **50. Business Insight**

A **Business Insight** is a meaningful interpretation of data or patterns that improves understanding of a business problem and can support action.

---

## **51. Business Decision**

A **Business Decision** is an action or choice made by management based on business objectives, available information, and analytical insights.

---

## **52. Actionable Insight**

An **Actionable Insight** is an analytical finding that can directly support or lead to a specific business action.

---

## **53. CRISP-DM**

**CRISP-DM** stands for **Cross-Industry Standard Process for Data Mining**.

It is a structured framework for conducting data mining projects through six phases:

1. Business Understanding
2. Data Understanding
3. Data Preparation
4. Modeling
5. Evaluation
6. Deployment

---

### **Descriptive → Diagnostic → Predictive → Prescriptive**

### **Business Problem → Data → Preparation → Exploration → Mining → Evaluation → Action**

### **Descriptive Mining → Patterns / Groups / Associations**

### **Predictive Mining → Classification / Prediction**
