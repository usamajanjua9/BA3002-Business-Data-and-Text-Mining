# **🟢 What is Data Collection?**

**Data Collection** is the systematic process of gathering information relevant to a specific question, problem, or objective.

### 📍 Proper Definition

> **Data Collection is the systematic process of gathering, recording, and organizing data from relevant sources for analysis and decision-making.**

---

## **🔹 Simple Example**

Suppose a restaurant wants to know:

> **Why are customers becoming less satisfied?**

It may collect:

* ⭐ Customer ratings
* 📝 Written reviews
* ⏱️ Delivery times
* 💰 Order amounts
* 🍔 Food categories
* 📍 Customer locations
* ☎️ Complaint records

The company is collecting data because it has a **specific business problem**.

---

# **🧠 Important Principle**

Do not start with:

> “What data can we collect?”

Start with:

> **“What business question are we trying to answer?”**

---

# **🟢 Business Problem Comes First**

### ❌ Weak Approach

> Let's collect everything.

### ✅ Better Approach

> We want to understand why customer satisfaction is declining.

Therefore, collect data related to:

* Ratings
* Complaints
* Delivery
* Product quality
* Customer experience

---

# **💼 Business Problem → Data Requirement**

| **Business Problem**     | **Possible Data Needed**                      |
| ------------------------ | --------------------------------------------- |
| Customers are leaving    | Complaints, usage, spending, contract history |
| Sales are decreasing     | Sales, products, prices, promotions           |
| Deliveries are late      | Order time, delivery time, location           |
| Fraud is increasing      | Transactions, amount, time, location          |
| Reviews are negative     | Ratings, review text, product information     |
| Inventory is running out | Sales history, inventory, demand              |

---

# **🟢 What is Data Acquisition?**

**Data Acquisition** refers to obtaining data from one or more sources and bringing it into a system where it can be stored, processed, or analyzed.

### 📍 Proper Definition

> **Data Acquisition is the process of obtaining data from different sources and transferring it into an environment where it can be used for analysis.**

---

## **🔹 Examples**

Data may be acquired by:

📁 Reading a CSV file

📗 Importing an Excel sheet

🗄️ Querying a database

🔌 Calling an API

🌐 Scraping a website

📋 Conducting a survey

---

# **🟢 Data Collection vs Data Acquisition**

The terms are closely related, but there is a useful distinction.

| **Data Collection**                      | **Data Acquisition**                                     |
| ---------------------------------------- | -------------------------------------------------------- |
| Broader process of gathering information | Focuses on obtaining data from a source                  |
| Starts from the information needed       | Focuses more on access and transfer                      |
| Can involve surveys and observation      | Often involves files, APIs, databases, sensors, websites |
| Concerned with *what* to gather          | Concerned with *how* to obtain it                        |

---

## **💡 Easy Way to Remember**

### Data Collection

> **What information do we need?**

### Data Acquisition

> **How do we get that information?**

---

# **🟢 Example**

A business wants competitor pricing information.

### Data Collection Decision

> We need competitor product names and prices.

### Data Acquisition Method

> Extract those values from an approved website or API.

---

# **🟢 Primary Data**

**Primary Data** is data collected directly for the specific purpose of the current analysis.

### 📍 Definition

> **Primary Data is original data collected directly by the researcher or organization for a specific problem or objective.**

---

## **Examples**

* Customer survey
* Interview
* Questionnaire
* Experiment
* Observation
* Customer feedback form

---

## **🔹 Business Example**

A company asks 1,000 customers:

> “How satisfied are you with our delivery service?”

The responses are **primary data** because the company collected them specifically for this question.

---

# **🟢 Secondary Data**

**Secondary Data** is data that already exists and was originally collected for another purpose.

### 📍 Definition

> **Secondary Data is existing data collected previously by another person, organization, or system and reused for the current analysis.**

---

## **Examples**

* Government statistics
* Company transaction records
* Existing reports
* Public datasets
* Kaggle datasets
* World Bank data
* Historical sales records

---

# **🟢 Primary vs Secondary Data**

| **Primary Data**                               | **Secondary Data**       |
| ---------------------------------------------- | ------------------------ |
| Collected specifically for the current problem | Already exists           |
| Usually more customized                        | Usually faster to obtain |
| Can be expensive                               | Often cheaper            |
| Survey                                         | Government dataset       |
| Interview                                      | Historical database      |
| Experiment                                     | Public dataset           |

---

# **💡 Which One is Better?**

Neither is automatically better.

It depends on the business problem.

A strong analytics project may combine:

# **Primary + Secondary Data**

---

# **🟢 Example**

Suppose a university wants to understand student satisfaction.

### Secondary Data

* Attendance
* Grades
* Course registrations

### Primary Data

* Student satisfaction survey
* Open-ended feedback

Combining both may provide a much stronger picture.

# **🟢 More Data ≠ Better Data**

A common misconception is:

> **“The more data we have, the better our analysis will be.”**

Not necessarily.

---

## **Consider This**

A company wants to predict customer churn.

It has:

* Monthly bill
* Complaints
* Usage
* Contract duration
* Payment history

These may be useful.

But it also has:

* Customer's favorite color
* Random internal record number
* File creation timestamp

These may add little or no useful information.

# **3. Common Business Data Formats**

---

## **📌 What is a Data Format?**

> **A Data Format is the structure in which data is stored, organized, or exchanged.**

The same business information can be stored in different formats.

### **Common Formats**

| Format   | Best For                        | Structure             |
| -------- | ------------------------------- | --------------------- |
| CSV      | Simple tabular data             | Rows & columns        |
| Excel    | Business reports and worksheets | Rows, columns, sheets |
| JSON     | APIs and web applications       | Keys and values       |
| Database | Large structured business data  | Related tables        |
| Text     | Reviews, comments, documents    | Unstructured          |

---

## **📄 CSV**

**CSV** stands for:

> **Comma-Separated Values**

Example:

```text
Customer_ID,City,Spending
C101,Islamabad,25000
C102,Lahore,18000
C103,Karachi,32000
```

### **Advantages**

* Simple
* Small
* Easy to share
* Works with most analytics tools

### **Limitation**

It does not support:

* Multiple sheets
* Formulas
* Rich formatting

---

## **📗 Excel**

Excel files usually use:

```text
.xlsx
```

They may contain:

* Multiple sheets
* Tables
* Formulas
* Charts
* Business reports

### **Example**

A company may maintain:

**Sheet 1:** Sales
**Sheet 2:** Customers
**Sheet 3:** Products

---

## **🧩 JSON**

JSON stands for:

> **JavaScript Object Notation**

It stores data using **keys and values**.

```json
{
  "customer_id": "C101",
  "city": "Islamabad",
  "spending": 25000
}
```

JSON is extremely common in:

* APIs
* Websites
* Mobile applications
* Modern software systems

---

## **🗄️ Database**

A database stores large amounts of organized data.

Example tables:

```text
CUSTOMERS
ORDERS
PRODUCTS
PAYMENTS
```

These tables can be connected using common fields such as:

```text
Customer_ID
Product_ID
Order_ID
```

---

## **💡 Easy Comparison**

### CSV

One simple table.

### Excel

Business spreadsheet with multiple sheets.

### JSON

Web/API-friendly structured information.

### Database

Large, connected business records.

---

# **🟢 SELECT and FROM**

Suppose we have a table called:

```text

```

To view all records:

```sql
SELECT *
FROM Orders;
```

`*` means:

> all columns

## **Select Specific Columns**

```sql
SELECT Customer_ID, Amount
FROM Orders;
```

This returns only the required variables.

# **🟢 Filtering with WHERE**

Management wants transactions above Rs. 10,000.

```sql
SELECT *
FROM Orders
WHERE Amount > 10000;
```

---

## **Business Question**

> Which customers returned their orders?

```sql
SELECT Customer_ID, Product_ID
FROM Orders
WHERE Returned = 'Yes';
```

---

# **🟢 Sorting with ORDER BY**

Find the largest transactions first:

```sql
SELECT *
FROM Orders
ORDER BY Amount DESC;
```

### `DESC`

Largest to smallest.

### `ASC`

Smallest to largest.

---

# **🟢 Aggregate Functions**

SQL can summarize data.

### Count Records

```sql
SELECT COUNT(*)
FROM Orders;
```

### Total Sales

```sql
SELECT SUM(Amount)
FROM Orders;
```

### Average Order Value

```sql
SELECT AVG(Amount)
FROM Orders;
```

Other useful functions:

```text

MAX()
COUNT()
SUM()
AVG()
```

# **🟢 GROUP BY**

Suppose management asks:

> What is the average order value for each city?

```sql
SELECT City, AVG(Amount)
FROM Orders
GROUP BY City;
```

Possible result:

| City      | Average Amount |
| --------- | -------------: |
| Islamabad |           8200 |
| Lahore    |           7100 |
| Karachi   |           9300 |

# **🟢 Simple Join**

```sql
SELECT Orders.Order_ID,
       Products.Product_Name,
       Orders.Amount
FROM Orders
JOIN Products
ON Orders.Product_ID = Products.Product_ID;
```

```python
import requests
from bs4 import BeautifulSoup
```

```python
title = soup.find("h2")
```

```python
products = soup.find_all("h2")
```


```python
price = soup.find(

)

```

Output:

```text
Rs. 2,500
```

```python
prices = soup.select(".price")
```

selects elements with:

```html
class="product"
```

# **17. Web Scraping vs APIs**

---

## **📌 Both Collect Web-Based Data**

But they work differently.

### **API**

The system intentionally provides data.

### **Web Scraping**

We extract information from webpage HTML.

---

# **🟢 Simple Comparison**

| **API**                        | **Web Scraping**                   |
| ------------------------------ | ---------------------------------- |
| Structured access              | Extracts webpage content           |
| Usually returns JSON           | Usually starts from HTML           |
| More stable                    | Can break when webpage changes     |
| Often documented               | Requires inspecting HTML           |
| May require API key            | May not require authentication     |
| Usually preferred if available | Useful when no suitable API exists |

---

# **🟢 Example**

QuickCart wants competitor prices.

### Option A — API Exists

```text
Competitor API
      ↓
JSON
      ↓
Pandas
```

✅ Prefer the API if it provides the required data under acceptable terms.

---

### Option B — No API Exists

```text
Competitor Website
       ↓
HTML
       ↓
BeautifulSoup
       ↓
DataFrame
```

Web scraping may be considered if appropriate and permitted.

---

# **🧠 Which Should You Choose?**

## **Choose an API when:**

* A suitable API exists
* Required data is available
* Documentation is clear
* Access is permitted
* Structured responses are useful

---

## **Consider Web Scraping when:**

* Required public information is visible on webpages
* No appropriate API exists
* Automated collection is permitted
* The volume and frequency are reasonable

# **📊 COMPLETE BASIC SCRAPER**

````python
# ============================================================
# BA3002 - BUSINESS DATA AND TEXT MINING
# WEEK 3 PRACTICAL
# WEB SCRAPING DEMO
# Business Case: BookWise Competitor Monitoring
# ============================================================


# ------------------------------------------------------------
# STEP 0: IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import requests                          # Used to send HTTP requests to websites.
from bs4 import BeautifulSoup            # Used to read, parse, and search HTML.
import pandas as pd                      # Used to create and analyze tabular data.


# ------------------------------------------------------------
# STEP 1: WEBSITE
# ------------------------------------------------------------

url = "https://books.toscrape.com/"      # Store the website address in a variable.


# ------------------------------------------------------------
# STEP 2: REQUEST THE WEBSITE
# ------------------------------------------------------------

response = requests.get(url, timeout=10) # Send a GET request to the website with a 10-second timeout.

response.encoding = "utf-8"              # Force UTF-8 encoding to correctly read symbols such as £.

print("Status Code:", response.status_code)  # Display the HTTP status code returned by the server.


# ------------------------------------------------------------
# STEP 3: CHECK WHETHER THE REQUEST WAS SUCCESSFUL
# ------------------------------------------------------------

if response.status_code == 200:          # Check whether the request succeeded.
    print("Website successfully retrieved.") # Inform the user that the webpage was downloaded.
else:                                    # Run this part if the website did not return status code 200.
    print("Website request failed.")      # Display a failure message.


# ------------------------------------------------------------
# STEP 4: PARSE THE HTML
# ------------------------------------------------------------

soup = BeautifulSoup(                    # Create a BeautifulSoup object from the webpage.
    response.text,                       # Use the HTML text received from the website.
    "html.parser"                        # Tell BeautifulSoup to use Python's HTML parser.
)


# ------------------------------------------------------------
# STEP 5: FIND ALL BOOK / PRODUCT CONTAINERS
# ------------------------------------------------------------

books = soup.find_all(                   # Search the webpage for all matching HTML elements.
    "article",                           # Look for <article> HTML tags.
    class_="product_pod"                 # Select only articles with class="product_pod".
)

print("Books Found:", len(books))         # Display how many book containers were found.


# ------------------------------------------------------------
# STEP 6: CREATE RATING MAPPING
# ------------------------------------------------------------

rating_map = {                           # Create a dictionary to convert rating words into numbers.
    "One": 1,                            # Convert One to numeric rating 1.
    "Two": 2,                            # Convert Two to numeric rating 2.
    "Three": 3,                          # Convert Three to numeric rating 3.
    "Four": 4,                           # Convert Four to numeric rating 4.
    "Five": 5                            # Convert Five to numeric rating 5.
}


# ------------------------------------------------------------
# STEP 7: CREATE EMPTY LISTS
# ------------------------------------------------------------

titles = []                              # Empty list for storing book titles.
prices = []                              # Empty list for storing book prices.
ratings = []                             # Empty list for storing book ratings.
availability_list = []                   # Empty list for storing availability information.


# ------------------------------------------------------------
# STEP 8: EXTRACT DATA FROM EVERY BOOK
# ------------------------------------------------------------

for book in books:                       # Loop through every book container found on the webpage.

    title = book.h3.a["title"]            # Extract the full book title from the title attribute.

    price = book.find(                   # Search inside the current book for its price.
        "p",                             # Look for a <p> HTML element.
        class_="price_color"             # Select the paragraph with class="price_color".
    ).get_text(strip=True)               # Extract only the text and remove extra spaces.

    availability = book.find(            # Search inside the current book for availability.
        "p",                             # Look for a <p> HTML element.
        class_="instock availability"    # Select the availability paragraph.
    ).get_text(strip=True)               # Extract availability text and remove extra spaces.

    rating_classes = book.find(          # Find the HTML element containing the star rating.
        "p",                             # Look for a <p> HTML element.
        class_="star-rating"             # Select the paragraph containing the rating class.
    )["class"]                            # Retrieve the list of HTML classes.

    rating_word = rating_classes[1]       # Take the second class, such as "Three" or "Five".

    rating = rating_map[rating_word]      # Convert the rating word into a numeric value.

    titles.append(title)                  # Add the extracted title to the titles list.
    prices.append(price)                  # Add the extracted price to the prices list.
    ratings.append(rating)                # Add the numeric rating to the ratings list.
    availability_list.append(availability) # Add availability information to its list.


# ------------------------------------------------------------
# STEP 9: CREATE A PANDAS DATAFRAME
# ------------------------------------------------------------

df = pd.DataFrame({                       # Create a DataFrame from the four collected lists.
    "Book_Title": titles,                 # Create the Book_Title column.
    "Price": prices,                      # Create the Price column.
    "Rating": ratings,                    # Create the Rating column.
    "Availability": availability_list     # Create the Availability column.
})


# ------------------------------------------------------------
# STEP 10: VIEW RAW SCRAPED DATA
# ------------------------------------------------------------

print("\nRaw Scraped Data:")              # Print a heading for the raw data.
print(df.head())                          # Display the first five scraped records.


# ------------------------------------------------------------
# STEP 11: CLEAN THE PRICE COLUMN
# ------------------------------------------------------------

df["Price"] = df["Price"].str.replace(    # Remove unwanted characters from the Price column.
    r"[^\d.]",                            # Keep only digits and the decimal point.
    "",                                  # Replace unwanted characters with nothing.
    regex=True                            # Treat the search pattern as a regular expression.
)

df["Price"] = pd.to_numeric(              # Convert cleaned price text into numeric values.
    df["Price"],                          # Specify the Price column to convert.
    errors="coerce"                       # Convert invalid values into NaN instead of crashing.
)


# ------------------------------------------------------------
# STEP 12: INSPECT THE CLEANED DATA
# ------------------------------------------------------------

print("\nFirst Five Cleaned Records:")    # Print a heading.
print(df.head())                          # Display the first five cleaned records.

print("\nDataset Shape:")                 # Print a heading for dataset size.
print(df.shape)                           # Display the number of rows and columns.

print("\nColumn Data Types:")             # Print a heading for data types.
print(df.dtypes)                          # Display the data type of every column.

print("\nMissing Values:")                # Print a heading for missing values.
print(df.isnull().sum())                  # Count missing values in every column.

print("\nDuplicates:")                    # Print a heading for duplicate records.
print(df.duplicated().sum())              # Count fully duplicated rows.


# ------------------------------------------------------------
# STEP 13: SUMMARY STATISTICS
# ------------------------------------------------------------

print("\nSummary Statistics:")            # Print a heading for descriptive statistics.
print(df.describe())                      # Display statistics for numeric columns.


# ------------------------------------------------------------
# STEP 14: CALCULATE AVERAGE PRICE
# ------------------------------------------------------------

average_price = df["Price"].mean()         # Calculate the mean price of all scraped books.

print(                                    # Start a print statement.
    "\nAverage Price:",                   # Display a label.
    round(average_price, 2)               # Round the average price to two decimal places.
)


# ------------------------------------------------------------
# STEP 15: CALCULATE MEDIAN PRICE
# ------------------------------------------------------------

median_price = df["Price"].median()        # Calculate the median book price.

print(                                    # Start a print statement.
    "Median Price:",                      # Display a label.
    round(median_price, 2)                # Round the median price to two decimal places.
)


# ------------------------------------------------------------
# STEP 16: CALCULATE AVERAGE RATING
# ------------------------------------------------------------

average_rating = df["Rating"].mean()       # Calculate the average star rating.

print(                                    # Start a print statement.
    "Average Rating:",                    # Display a label.
    round(average_rating, 2)              # Round the average rating to two decimal places.
)


# ------------------------------------------------------------
# STEP 17: RATING DISTRIBUTION
# ------------------------------------------------------------

rating_distribution = (                   # Create a variable to store rating counts.
    df["Rating"]                          # Select the Rating column.
    .value_counts()                       # Count how many books have each rating.
    .sort_index()                         # Sort ratings from 1 star to 5 stars.
)

print("\nRating Distribution:")           # Print a heading for rating distribution.
print(rating_distribution)                # Display the number of books in each rating group.


# ------------------------------------------------------------
# STEP 18: FIND THE MOST EXPENSIVE BOOK
# ------------------------------------------------------------

most_expensive = df.loc[                  # Select the full row of the most expensive book.
    df["Price"].idxmax()                  # Find the row index containing the maximum price.
]

print("\nMost Expensive Book:")           # Print a heading.
print(most_expensive)                     # Display information about the most expensive book.


# ------------------------------------------------------------
# STEP 19: FIND THE LEAST EXPENSIVE BOOK
# ------------------------------------------------------------

least_expensive = df.loc[                 # Select the full row of the least expensive book.
    df["Price"].idxmin()                  # Find the row index containing the minimum price.
]

print("\nLeast Expensive Book:")          # Print a heading.
print(least_expensive)                    # Display information about the least expensive book.


# ------------------------------------------------------------
# STEP 20: FIND HIGH-RATED PRODUCTS
# ------------------------------------------------------------

high_rated = df[                          # Create a filtered DataFrame.
    df["Rating"] >= 4                     # Keep only books with ratings of 4 or 5 stars.
]

print("\nHighly Rated Books:")            # Print a heading.
print(high_rated)                         # Display all highly rated books.


# ------------------------------------------------------------
# STEP 21: FIND LOWER-PRICED AND HIGH-RATED BOOKS
# ------------------------------------------------------------

opportunities = df[                       # Create another filtered DataFrame.
    (df["Rating"] >= 4) &                 # Require a rating of at least 4 stars.
    (df["Price"] <= median_price)         # Require price to be at or below the median price.
]

print("\nPotential Opportunities:")       # Print a heading.
print(opportunities)                      # Display potentially attractive books.


# ------------------------------------------------------------
# STEP 22: CALCULATE AVERAGE PRICE BY RATING
# ------------------------------------------------------------

price_by_rating = (                       # Create a summary table.
    df.groupby("Rating")["Price"]          # Group book prices according to star rating.
    .mean()                               # Calculate the average price for each rating group.
    .round(2)                             # Round values to two decimal places.
)

print("\nAverage Price by Rating:")       # Print a heading.
print(price_by_rating)                    # Display average price for each rating.


# ------------------------------------------------------------
# STEP 23: SAVE THE SCRAPED DATA
# ------------------------------------------------------------

df.to_csv(                                # Export the DataFrame to a CSV file.
    "bookwise_competitor_data.csv",       # Specify the output filename.
    index=False                           # Do not save Pandas row numbers as a column.
)

print(                                    # Start a final print statement.
    "\nDataset successfully saved as "    # Display confirmation text.
    "'bookwise_competitor_data.csv'."     # Display the output filename.
)


# ------------------------------------------------------------
# STEP 24: FINAL BUSINESS SUMMARY
# ------------------------------------------------------------

print("\n--- BUSINESS SUMMARY ---")       # Print a heading for the final summary.

print("Books Collected:", len(df))         # Display total number of collected books.

print(                                    # Start a print statement.
    "Average Listed Price:",              # Display a label.
    round(df["Price"].mean(), 2)          # Display the average price.
)

print(                                    # Start a print statement.
    "Median Listed Price:",               # Display a label.
    round(df["Price"].median(), 2)        # Display the median price.
)

print(                                    # Start a print statement.
    "Books Rated 4 or Higher:",           # Display a label.
    len(high_rated)                       # Count books with ratings of 4 or 5.
)

print(                                    # Start a print statement.
    "Potential Opportunities:",           # Display a label.
    len(opportunities)                    # Count high-rated books at or below median price.
)
````

# **STEP 34 — Scrape Three Pages**

For classroom purposes, collect only:

# **3 Pages**

approximately:

This is enough to demonstrate pagination without sending unnecessary requests.

---



````python
# ============================================================
# BA3002 - BUSINESS DATA AND TEXT MINING
# WEEK 3 PRACTICAL
# MULTI-PAGE WEB SCRAPING DEMO
# Business Case: BookWise Competitor Monitoring
# ============================================================


# ------------------------------------------------------------
# STEP 0: IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import requests                          # Used to send HTTP requests to websites.
from bs4 import BeautifulSoup            # Used to parse and search HTML content.
import pandas as pd                      # Used to create and analyze tabular datasets.
import time                              # Used to pause between requests for responsible scraping.


# ------------------------------------------------------------
# STEP 1: CREATE EMPTY LISTS
# ------------------------------------------------------------

titles = []                              # Store book titles.
prices = []                              # Store book prices.
ratings = []                             # Store numeric star ratings.
availability_list = []                   # Store availability information.
page_numbers = []                        # Store the page number from which each book was collected.


# ------------------------------------------------------------
# STEP 2: CREATE RATING MAPPING
# ------------------------------------------------------------

rating_map = {                           # Convert rating words from HTML into numbers.
    "One": 1,                            # One star becomes 1.
    "Two": 2,                            # Two stars become 2.
    "Three": 3,                          # Three stars become 3.
    "Four": 4,                           # Four stars become 4.
    "Five": 5                            # Five stars become 5.
}


# ------------------------------------------------------------
# STEP 3: LOOP THROUGH MULTIPLE WEBPAGES
# ------------------------------------------------------------

for page in range(1, 4):                 # Loop through page 1, page 2, and page 3.


    # --------------------------------------------------------
    # STEP 3.1: CREATE THE URL FOR THE CURRENT PAGE
    # --------------------------------------------------------

    url = (                              # Start building the webpage URL.
        "https://books.toscrape.com/"    # Base address of the website.
        f"catalogue/page-{page}.html"    # Add the current page number to the URL.
    )


    # Display which page is currently being processed.
    print("\nScraping:", url)


    # --------------------------------------------------------
    # STEP 3.2: SEND THE HTTP REQUEST
    # --------------------------------------------------------

    try:                                 # Start error handling for the web request.

        response = requests.get(         # Send a GET request to the current page.
            url,                         # Website address to request.
            timeout=10                   # Stop waiting after 10 seconds if no response arrives.
        )

        response.encoding = "utf-8"      # Force UTF-8 to correctly read symbols such as £.

        response.raise_for_status()      # Raise an error if the server returns an unsuccessful status.


    except requests.RequestException as error:  # Run if the request fails.

        print(                           # Display an informative error message.
            "Page failed:",              # Explain that this page could not be retrieved.
            page,                        # Display the failed page number.
            "-",                         # Separator.
            error                        # Display the actual request error.
        )

        continue                         # Skip this page and move to the next one.


    # Display the HTTP status code for successful requests.
    print("Status Code:", response.status_code)


    # --------------------------------------------------------
    # STEP 3.3: PARSE THE HTML
    # --------------------------------------------------------

    soup = BeautifulSoup(                # Create a BeautifulSoup object.
        response.text,                   # Use the HTML returned by the server.
        "html.parser"                    # Parse the content as HTML.
    )


    # --------------------------------------------------------
    # STEP 3.4: FIND ALL BOOK CONTAINERS
    # --------------------------------------------------------

    books = soup.find_all(               # Find all book/product containers.
        "article",                       # Look for <article> tags.
        class_="product_pod"             # Select only elements with class="product_pod".
    )


    # Display how many books were found on this page.
    print("Books Found:", len(books))


    # --------------------------------------------------------
    # STEP 3.5: LOOP THROUGH EVERY BOOK ON THE PAGE
    # --------------------------------------------------------

    for book in books:                   # Process each book one at a time.


        # ----------------------------------------------------
        # EXTRACT BOOK TITLE
        # ----------------------------------------------------

        title = book.h3.a["title"]        # Extract the full title from the title attribute.


        # ----------------------------------------------------
        # EXTRACT PRICE
        # ----------------------------------------------------

        price = book.find(               # Search inside the current book container.
            "p",                         # Look for a paragraph element.
            class_="price_color"         # Select the paragraph containing the price.
        ).get_text(strip=True)           # Extract only the price text and remove extra spaces.


        # ----------------------------------------------------
        # EXTRACT AVAILABILITY
        # ----------------------------------------------------

        availability = book.find(        # Search for availability information.
            "p",                         # Look for a paragraph element.
            class_="instock availability" # Select the availability paragraph.
        ).get_text(strip=True)           # Extract the text and remove unnecessary spaces.


        # ----------------------------------------------------
        # EXTRACT STAR RATING
        # ----------------------------------------------------

        rating_classes = book.find(      # Locate the element containing the star rating.
            "p",                         # Look for a paragraph element.
            class_="star-rating"         # Select the star-rating paragraph.
        )["class"]                        # Retrieve all CSS classes attached to the element.


        rating_word = rating_classes[1]  # Get the rating word such as "Three" or "Five".


        rating = rating_map[             # Use the dictionary to convert the word into a number.
            rating_word                  # Rating word used as the dictionary key.
        ]


        # ----------------------------------------------------
        # STORE EXTRACTED VALUES
        # ----------------------------------------------------

        titles.append(title)             # Add the book title to the titles list.

        prices.append(price)             # Add the raw price to the prices list.

        ratings.append(rating)           # Add the numeric rating to the ratings list.

        availability_list.append(        # Add availability information to the list.
            availability
        )

        page_numbers.append(page)        # Record the page from which this book was collected.


    # --------------------------------------------------------
    # STEP 3.6: RESPONSIBLE SCRAPING DELAY
    # --------------------------------------------------------

    print("Pausing before next page...") # Inform the user that the script will pause.

    time.sleep(1)                        # Wait one second before requesting the next page.


# ------------------------------------------------------------
# STEP 4: CREATE ONE DATAFRAME FROM ALL SCRAPED PAGES
# ------------------------------------------------------------

df = pd.DataFrame({                      # Create a Pandas DataFrame.
    "Book_Title": titles,                # Add titles as one column.
    "Price": prices,                     # Add prices as one column.
    "Rating": ratings,                   # Add ratings as one column.
    "Availability": availability_list,   # Add availability as one column.
    "Page": page_numbers                 # Add source page number as one column.
})


# ------------------------------------------------------------
# STEP 5: VIEW THE RAW SCRAPED DATA
# ------------------------------------------------------------

print("\nRaw Scraped Data:")             # Print a heading.

print(df.head())                         # Display the first five records.


# ------------------------------------------------------------
# STEP 6: CLEAN THE PRICE COLUMN
# ------------------------------------------------------------

df["Price"] = df["Price"].str.replace(   # Clean unwanted characters from Price.
    r"[^\d.]",                           # Keep only digits and decimal points.
    "",                                 # Replace unwanted characters with nothing.
    regex=True                           # Treat the pattern as a regular expression.
)


df["Price"] = pd.to_numeric(             # Convert the cleaned price values to numbers.
    df["Price"],                         # Select the Price column.
    errors="coerce"                      # Convert problematic values to NaN rather than crashing.
)


# ------------------------------------------------------------
# STEP 7: INSPECT THE FINAL DATASET
# ------------------------------------------------------------

print("\nFirst Five Cleaned Records:")   # Print a heading.

print(df.head())                         # Display the first five cleaned records.


print("\nDataset Shape:")                # Print a heading.

print(df.shape)                          # Display rows and columns.


print("\nColumn Data Types:")            # Print a heading.

print(df.dtypes)                         # Display the data type of each variable.


print("\nMissing Values:")               # Print a heading.

print(df.isnull().sum())                 # Count missing values in each column.


print("\nDuplicate Records:")            # Print a heading.

print(df.duplicated().sum())             # Count fully duplicated rows.


# ------------------------------------------------------------
# STEP 8: SUMMARY STATISTICS
# ------------------------------------------------------------

print("\nSummary Statistics:")           # Print a heading.

print(df.describe())                     # Display descriptive statistics for numeric variables.


# ------------------------------------------------------------
# STEP 9: SAVE THE COMPLETE SCRAPED DATASET
# ------------------------------------------------------------

df.to_csv(                               # Export the DataFrame to a CSV file.
    "bookwise_multi_page_data.csv",      # Name of the output file.
    index=False                          # Do not save Pandas row numbers.
)


print(                                   # Print confirmation.
    "\nDataset successfully saved as "
    "'bookwise_multi_page_data.csv'."
)
````

# BEGINNER TUTORIAL: USING API DATA WITH PYTHON

````python
# ============================================================
# BEGINNER TUTORIAL: USING API DATA WITH PYTHON
# ============================================================

# Import the requests library.
# This library is used to send requests to websites and APIs.
import requests

# Import the pandas library.
# Pandas is used to organize and analyze data in table form.
import pandas as pd


# ============================================================
# STEP 1: DEFINE THE API URL
# ============================================================

# Store the API endpoint inside a variable called url.
# An endpoint is the specific API address from which we request data.
# JSONPlaceholder is a free practice API that does not require an API key.
url = "https://jsonplaceholder.typicode.com/users"


# ============================================================
# STEP 2: SEND A GET REQUEST TO THE API
# ============================================================

# Send a GET request to the API.
# GET is used when we want to read or download data.
response = requests.get(url)


# ============================================================
# STEP 3: CHECK THE HTTP STATUS CODE
# ============================================================

# Print the status code returned by the API server.
# Status code 200 means that the request was successful.
print("Status Code:", response.status_code)


# ============================================================
# STEP 4: CHECK WHETHER THE REQUEST WAS SUCCESSFUL
# ============================================================

# Check whether the API returned status code 200.
if response.status_code == 200:

    # Display a success message if the request was successful.
    print("API request was successful.")

else:

    # Display an error message if the API request failed.
    print("API request failed.")

    # Stop the program because there is no valid data to process.
    exit()


# ============================================================
# STEP 5: CONVERT THE API RESPONSE INTO JSON
# ============================================================

# Convert the API response from JSON format into Python data.
# JSON data usually becomes a Python list or dictionary.
data = response.json()


# ============================================================
# STEP 6: CHECK THE TYPE OF DATA RECEIVED
# ============================================================

# Print the Python data type of the returned API data.
print("\nType of API Data:")

# Display the data type.
print(type(data))


# ============================================================
# STEP 7: CHECK HOW MANY RECORDS WERE RECEIVED
# ============================================================

# Count the number of records returned by the API.
number_of_records = len(data)

# Display the total number of records.
print("\nTotal Records Received:", number_of_records)


# ============================================================
# STEP 8: VIEW THE FIRST RECORD
# ============================================================

# Get the first record from the API data.
# Python starts counting from 0, so data[0] means the first record.
first_user = data[0]

# Display a heading.
print("\nFirst Record:")

# Print the complete first user record.
print(first_user)


# ============================================================
# STEP 9: ACCESS INDIVIDUAL VALUES FROM THE FIRST RECORD
# ============================================================

# Get the ID of the first user.
user_id = first_user["id"]

# Get the name of the first user.
user_name = first_user["name"]

# Get the username of the first user.
username = first_user["username"]

# Get the email address of the first user.
email = first_user["email"]


# Display a heading.
print("\nSelected Information from First Record:")

# Display the user's ID.
print("ID:", user_id)

# Display the user's name.
print("Name:", user_name)

# Display the username.
print("Username:", username)

# Display the user's email address.
print("Email:", email)


# ============================================================
# STEP 10: LOOP THROUGH ALL API RECORDS
# ============================================================

# Display a heading before printing all users.
print("\nAll Users:")

# Start a loop that processes every user in the API data.
for user in data:

    # Get the ID of the current user.
    current_id = user["id"]

    # Get the name of the current user.
    current_name = user["name"]

    # Get the email address of the current user.
    current_email = user["email"]

    # Display the current user's information.
    print(current_id, "-", current_name, "-", current_email)


# ============================================================
# STEP 11: CONVERT JSON DATA INTO A PANDAS DATAFRAME
# ============================================================

# Convert the JSON data into a Pandas DataFrame.
# A DataFrame stores data in rows and columns like Excel.
df = pd.DataFrame(data)


# ============================================================
# STEP 12: VIEW THE COMPLETE DATAFRAME
# ============================================================

# Display a heading.
print("\nComplete DataFrame:")

# Print the DataFrame.
print(df)


# ============================================================
# STEP 13: VIEW THE FIRST FIVE ROWS
# ============================================================

# Display a heading.
print("\nFirst Five Rows:")

# Display the first five rows of the DataFrame.
print(df.head())


# ============================================================
# STEP 14: CHECK THE COLUMN NAMES
# ============================================================

# Display a heading.
print("\nColumn Names:")

# Display all available column names.
print(df.columns)


# ============================================================
# STEP 15: SELECT ONLY IMPORTANT COLUMNS
# ============================================================

# Select only the columns that we want to keep.
# This removes unnecessary nested columns such as address and company.
df = df[
    [
        "id",
        "name",
        "username",
        "email",
        "phone",
        "website"
    ]
]


# ============================================================
# STEP 16: DISPLAY THE CLEAN DATA
# ============================================================

# Display a heading.
print("\nSelected API Data:")

# Display the cleaned DataFrame.
print(df)


# ============================================================
# STEP 17: CHECK THE SIZE OF THE DATASET
# ============================================================

# Get the number of rows in the DataFrame.
number_of_rows = df.shape[0]

# Get the number of columns in the DataFrame.
number_of_columns = df.shape[1]

# Display the total number of rows.
print("\nNumber of Rows:", number_of_rows)

# Display the total number of columns.
print("Number of Columns:", number_of_columns)


# ============================================================
# STEP 18: CHECK DATA TYPES
# ============================================================

# Display a heading.
print("\nData Types:")

# Display the data type of each column.
print(df.dtypes)


# ============================================================
# STEP 19: CHECK FOR MISSING VALUES
# ============================================================

# Display a heading.
print("\nMissing Values:")

# Count missing values in every column.
print(df.isnull().sum())


# ============================================================
# STEP 20: CHECK FOR DUPLICATE RECORDS
# ============================================================

# Count the number of duplicate rows.
duplicate_records = df.duplicated().sum()

# Display the number of duplicate records.
print("\nDuplicate Records:", duplicate_records)


# ============================================================
# STEP 21: REMOVE DUPLICATES
# ============================================================

# Remove duplicate rows if any duplicates exist.
df = df.drop_duplicates()


# ============================================================
# STEP 22: SORT THE DATA
# ============================================================

# Sort the users alphabetically according to their names.
df = df.sort_values(
    by="name"
)


# ============================================================
# STEP 23: RESET THE DATAFRAME INDEX
# ============================================================

# Reset the row numbers after sorting.
# drop=True prevents the old index from becoming a new column.
df = df.reset_index(
    drop=True
)


# ============================================================
# STEP 24: DISPLAY THE FINAL CLEAN DATASET
# ============================================================

# Display a heading.
print("\nFinal Clean API Dataset:")

# Display the final cleaned and sorted DataFrame.
print(df)


# ============================================================
# STEP 25: SAVE THE API DATA AS A CSV FILE
# ============================================================

# Store the output file name in a variable.
file_name = "users_api_data.csv"

# Save the DataFrame as a CSV file.
# index=False prevents Pandas from saving row numbers as a separate column.
df.to_csv(
    file_name,
    index=False
)


# ============================================================
# STEP 26: CONFIRM THAT THE FILE WAS SAVED
# ============================================================

# Display a confirmation message.
print("\nData successfully saved as:", file_name)


# ============================================================
# FINAL SUMMARY
# ============================================================

# Display a heading.
print("\nAPI DATA COLLECTION COMPLETED")

# Display the basic workflow used in this program.
print("API URL -> GET Request -> JSON -> DataFrame -> Cleaning -> CSV")
````
