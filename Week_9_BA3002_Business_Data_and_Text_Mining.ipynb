# 📌 **1.3 Association vs Clustering vs Classification**

---

## 🔥 Comparison Table

| Feature       | Association        | Clustering      | Classification    |
| ------------- | ------------------ | --------------- | ----------------- |
| Goal          | Find relationships | Group data      | Predict category  |
| Learning Type | Unsupervised       | Unsupervised    | Supervised        |
| Output        | Rules (A → B)      | Clusters        | Labels            |
| Example       | Bread → Butter     | Customer groups | Fraud / Not Fraud |

---

---

## 🎯 Easy Understanding

👉 **Association**

* Finds relationships
* Example:

```text
Bread → Butter
```

---

👉 **Clustering**

* Groups similar data
* Example:
* High spenders / Low spenders

---

👉 **Classification**

* Predicts category
* Example:
* Fraud / Not Fraud

---

---

## 🧠 Memory Trick:

👉 Relationship → Association
👉 Grouping → Clustering
👉 Prediction → Classification

# 🔹 **3. Association Rules & Metrics (Core Concept + Math Combined)**

---

# 📌 **3.1 What is an Association Rule?**

---

## ✅ **Definition**

👉 An **association rule** shows a relationship between items

---

## 💡 In simple words:

> If a customer buys item A, they are likely to buy item B

---

## 🎯 Example:

```text
Bread → Butter
```

👉 Meaning:

* Customers who buy Bread
* often also buy Butter

---

## 🧠 Key Idea:

* Not prediction
* Not grouping
  👉 It is **relationship discovery**

---

---

# 📌 **3.2 Rule Format (A → B)**

---

## ✅ General Format:

```text
A → B
```

---

## 🎯 Meaning:

> If A happens → B is likely to happen

---

## 📦 Example:

```text
Milk → Bread
```

👉 If a customer buys Milk
👉 they are likely to buy Bread

---

---

# 📌 **3.3 Antecedent (LHS) and Consequent (RHS)**

---

## ✅ Definitions:

* **Antecedent (LHS)** → Left side (A)
* **Consequent (RHS)** → Right side (B)

---

## 🎯 Example:

```text
Bread → Butter
```

| Part   | Meaning          |
| ------ | ---------------- |
| Bread  | Antecedent (LHS) |
| Butter | Consequent (RHS) |

---

---

## 🧠 Simple Understanding:

👉 “IF part” → Antecedent

👉 “THEN part” → Consequent

---

---

# 📌 **Dataset (Used for All Calculations)**

---

| Transaction | Items               |
| ----------- | ------------------- |
| T1          | Bread, Milk         |
| T2          | Bread, Butter       |
| T3          | Milk, Butter        |
| T4          | Bread, Milk, Butter |
| T5          | Bread, Milk         |

---

👉 Total Transactions:

```text
Total = 5
```

---

---

# 📌 **3.4 Support (Definition + Formula + Example)**

---

## ✅ **Definition**

👉 Support measures **how frequently items appear together**

---

## 📌 Formula:

$$
Support(A \rightarrow B) = \frac{\text{Number of transactions containing A and B}}{\text{Total transactions}}
$$

---

## 🎯 Example:

Rule:

```text
Bread → Milk
```

---

### Step 1: Count transactions containing BOTH Bread and Milk

* T1 → Yes
* T4 → Yes
* T5 → Yes

👉 Count = 3

---

### Step 2: Total transactions

👉 Total = 5

---

### Step 3: Apply formula

$$
Support(Bread \rightarrow Milk) = \frac{3}{5}
$$

$$
Support = 0.6
$$

---

## 🧠 Interpretation:

👉 60% of transactions contain both Bread and Milk

---

---

# 📌 **3.5 Confidence (Definition + Formula + Example)**

---

## ✅ **Definition**

👉 Confidence measures **how often B is bought when A is bought**

---

## 📌 Formula:

$$
Confidence(A \rightarrow B) = \frac{\text{Number of transactions containing A and B}}{\text{Number of transactions containing A}}
$$

---

## 🎯 Example:

Rule:

```text
Bread → Milk
```

---

### Step 1: Transactions with Bread

T1, T2, T4, T5


👉 Count = 4

---

### Step 2: Transactions with Bread AND Milk

👉 Count = 3

---

### Step 3: Apply formula

$$
Confidence(Bread \rightarrow Milk) = \frac{3}{4}
$$

$$
Confidence = 0.75
$$

---

## 🧠 Interpretation:

👉 75% of customers who buy Bread also buy Milk

---

---

# 📌 **3.6 Lift (Definition + Formula + Example)**

---

## ✅ **Definition**

👉 Lift measures **how strong the relationship is compared to random chance**

---

## 📌 Formula:

$$
Lift(A \rightarrow B) = \frac{Confidence(A \rightarrow B)}{Support(B)}
$$

---

---

## 🎯 Example:

Rule:

```text
Bread → Milk
```

---

### Step 1: Confidence

$$
Confidence = 0.75
$$

---

### Step 2: Support of Milk

Milk appears in:

* T1, T3, T4, T5
  👉 Count = 4

$$
Support(Milk) = \frac{4}{5} = 0.8
$$

---

### Step 3: Apply formula

$$
Lift = \frac{0.75}{0.8}
$$

$$
Lift = 0.94
$$

---

## 🧠 Interpretation:

* Lift > 1 → strong positive relationship
* Lift = 1 → no relationship
* Lift < 1 → weak or negative relationship

---

👉 Here:

```text
Lift = 0.94 → weak relationship
```

---

---

# 📌 **3.7 Interpreting Rules Using Metrics (Strong vs Weak Rules)**

---

## 🔥 How to Evaluate a Rule

---

### ✅ Good Rule:

* High Support
* High Confidence
* Lift > 1

---

---

### ❌ Weak Rule:

* Low support
* Low confidence
* Lift ≤ 1

---

---

## 🎯 Example Summary

| Metric     | Value | Meaning            |
| ---------- | ----- | ------------------ |
| Support    | 0.6   | Common combination |
| Confidence | 0.75  | Strong likelihood  |
| Lift       | 0.94  | Weak relationship  |

---

---

## 🧠 Final Interpretation:

👉 Even though:

* support is good
* confidence is good

👉 Lift tells us:

❗ Relationship is not very strong

---

---

# 🎯 **Final Understanding**

👉 Association rules are evaluated using:

* Support → how common
* Confidence → how reliable
* Lift → how strong

---

---

# ✅ **Quick Recap**

* Rule format:

```text
A → B
```

* Metrics:

  * Support
  * Confidence
  * Lift

# 🔹 **4. Apriori Algorithm (Core Algorithm)**

---

# 📌 **4.1 What is Apriori Algorithm?**

---

## ✅ **Definition**

👉 **Apriori Algorithm is a method used to find items that frequently occur together in transaction data.**

---

## 💡 In simple words:

> Apriori helps us discover **common combinations of items** in customer purchases

---

## 🎯 Example:

Given transactions:

* Bread, Milk
* Bread, Butter
* Milk, Butter

👉 Apriori finds patterns like:

```text
Bread + Milk (frequent)
Bread + Butter (frequent)
```

---

👉 Then generates rules like:

```text
Bread → Milk
Bread → Butter
```

---

---

## 🧠 Key Idea:

👉 Apriori works in **two main steps**:

1. Find **frequent itemsets**
2. Generate **association rules**

---

---

## 🔍 Where is it used?

* Retail (shopping baskets)
* E-commerce recommendations
* Banking product combinations
* Marketing strategies

---

---

## 🧠 Real-Life Analogy

👉 Think like a shopkeeper:

> “Which items do customers usually buy together?”

---

---

# 📌 **4.2 Key Idea: Frequent Itemsets**

---

## ✅ **What is an Itemset?**

👉 A group of one or more items

---

### 🎯 Examples:

```text
{Bread}
{Milk}
{Bread, Milk}
{Bread, Milk, Butter}
```

---

---

## ✅ **What is a Frequent Itemset?**

👉 An itemset that appears **many times in transactions**

---

## 🎯 Example:

If:

```text
Bread + Milk
```

appears in 3 out of 5 transactions

👉 It is a **frequent itemset**

---

---

## 🧠 Important Concept: Support

👉 Frequency is measured using **Support**

$$
Support = \frac{\text{Number of transactions containing itemset}}{\text{Total transactions}}
$$

---

---

## 🔥 Core Principle of Apriori (VERY IMPORTANT)

---

## 👉 Apriori Property:

> If an itemset is frequent, then all of its subsets must also be frequent

---

---

## 🎯 Example:

If:

```text
{Bread, Milk, Butter}
```

is frequent

👉 Then these must also be frequent:

```text
{Bread, Milk}
{Bread, Butter}
{Milk, Butter}
```

---

---

## ❗ Important Rule:

> If a smaller itemset is NOT frequent, then the larger itemset CANNOT be frequent

---

---

## 🎯 Example:

If:

```text
Milk + Butter = NOT frequent
```

👉 Then:

```text
Bread + Milk + Butter = NOT possible
```

---

---

## 🧠 Why this is important?

👉 It helps:

* reduce calculations
* remove useless combinations
* make algorithm efficient

---

---

## 🎯 Visual Understanding

```text
Step 1: Find frequent single items
Step 2: Combine into pairs
Step 3: Remove weak pairs
Step 4: Build larger sets only from strong ones
```

---


# 📘 Complete Apriori Algorithm

## 🎯 Goal

We want to find:

> Which products are commonly bought together?

We will use the same type of example shown in the image.

---

# 🔹 Step 1: Transaction Data

## 📦 Market Basket Dataset

| TID | Items Purchased     |
| --- | ------------------- |
| T1  | Bread, Milk         |
| T2  | Bread, Butter       |
| T3  | Milk, Butter        |
| T4  | Bread, Milk, Butter |
| T5  | Bread, Milk         |
| T6  | Bread, Butter       |

---

## 🧠 Meaning

Each row is one customer purchase.

For example:

* T1 means one customer bought Bread and Milk.
* T4 means one customer bought Bread, Milk, and Butter.

Total transactions:

$$
N = 6
$$

---

# 🔹 Step 2: Give Short Names to Items

To make calculations easier:

| Item   | Short Name |
| ------ | ---------- |
| Bread  | B          |
| Milk   | M          |
| Butter | Bu         |

---

# 🔹 Step 3: Set Minimum Support

Assume:

$$
Minimum\ Support = 0.50
$$

This means an itemset must appear in at least 50% of transactions.

Since there are 6 transactions:

$$
Minimum\ Support\ Count = 0.50 \times 6
$$

$$
Minimum\ Support\ Count = 3
$$

So, an item or item pair must appear at least **3 times** to be called frequent.

---

# 🔹 Step 4: Count Single Items

Now count each item.

| Item        | Transactions       | Count |
| ----------- | ------------------ | ----: |
| Bread (B)   | T1, T2, T4, T5, T6 |     5 |
| Milk (M)    | T1, T3, T4, T5     |     4 |
| Butter (Bu) | T2, T3, T4, T6     |     4 |

---

# 🔹 Step 5: Calculate Support for Single Items

## 📌 Formula

$$
Support(Item) = \frac{Number\ of\ transactions\ containing\ item}{Total\ transactions}
$$

---

## Bread

$$
Support(B) = \frac{5}{6}
$$

$$
Support(B) = 0.83
$$

---

## Milk

$$
Support(M) = \frac{4}{6}
$$

$$
Support(M) = 0.67
$$

---

## Butter

$$
Support(Bu) = \frac{4}{6}
$$

$$
Support(Bu) = 0.67
$$

---

## ✅ Result

| Item | Count | Support | Frequent? |
| ---- | ----: | ------: | --------- |
| B    |     5 |    0.83 | Yes       |
| M    |     4 |    0.67 | Yes       |
| Bu   |     4 |    0.67 | Yes       |

All single items are frequent because all supports are greater than or equal to:

$$
0.50
$$

---

# 🔹 Step 6: Generate Candidate Item Pairs

Because all single items are frequent, we now create pairs.

Possible pairs:

| Candidate Pair |
| -------------- |
| B, M           |
| B, Bu          |
| M, Bu          |

This step is called **candidate generation**.

---

# 🔹 Step 7: Count Item Pairs

Now count how many times each pair appears together.

---

## Pair: Bread + Milk

Appears in:

* T1: Bread, Milk
* T4: Bread, Milk, Butter
* T5: Bread, Milk

Count:

$$
Count(B,M) = 3
$$

---

## Pair: Bread + Butter

Appears in:

* T2: Bread, Butter
* T4: Bread, Milk, Butter
* T6: Bread, Butter

Count:

$$
Count(B,Bu) = 3
$$

---

## Pair: Milk + Butter

Appears in:

* T3: Milk, Butter
* T4: Bread, Milk, Butter

Count:

$$
Count(M,Bu) = 2
$$

---

# 🔹 Step 8: Calculate Support for Item Pairs

## 📌 Formula

$$
Support(A,B) = \frac{Number\ of\ transactions\ containing\ both\ A\ and\ B}{Total\ transactions}
$$

---

## Bread + Milk

$$
Support(B,M) = \frac{3}{6}
$$

$$
Support(B,M) = 0.50
$$

---

## Bread + Butter

$$
Support(B,Bu) = \frac{3}{6}
$$

$$
Support(B,Bu) = 0.50
$$

---

## Milk + Butter

$$
Support(M,Bu) = \frac{2}{6}
$$

$$
Support(M,Bu) = 0.33
$$

---

## ✅ Result

| Item Pair | Count | Support | Frequent? |
| --------- | ----: | ------: | --------- |
| B, M      |     3 |    0.50 | Yes       |
| B, Bu     |     3 |    0.50 | Yes       |
| M, Bu     |     2 |    0.33 | No        |

---

# 🔹 Step 9: Pruning

## ✅ What is Pruning?

Pruning means removing itemsets that do not meet minimum support.

Minimum support:

$$
0.50
$$

So:

$$
Support(M,Bu) = 0.33 < 0.50
$$

Therefore:

$$
(M,Bu)\ is\ removed
$$

---

## ✅ Frequent Pairs Remaining

| Frequent Pair |
| ------------- |
| B, M          |
| B, Bu         |

---

# 🔹 Step 10: Try to Generate Triple Itemset

Now we check whether the triple exists:

$$
(B, M, Bu)
$$

It appears in:

* T4 only

So:

$$
Count(B,M,Bu) = 1
$$

Support:

$$
Support(B,M,Bu) = \frac{1}{6}
$$

$$
Support(B,M,Bu) = 0.17
$$

Since:

$$
0.17 < 0.50
$$

the triple is not frequent.

---

## 🧠 Apriori Logic

Also, one of its smaller pairs was not frequent:

$$
(M,Bu)
$$

So the bigger set:

$$
(B,M,Bu)
$$

cannot be frequent.

This is the main Apriori idea:

> If a smaller itemset is not frequent, the larger itemset cannot be frequent.

---

# 🔹 Step 11: Stop

We stop because no frequent triple exists.

Final frequent itemsets are:

## Frequent 1-itemsets


{B}, {M}, {Bu}


## Frequent 2-itemsets


{B,M}, {B,Bu}


---

# 🔹 Step 12: Generate Association Rules

Rules are generated from frequent pairs.

From:


{B,M}


we get:

```text
B → M
M → B
```

From:


{B,Bu}


we get:

```text
B → Bu
Bu → B
```

So total rules are:

| Rule   | Meaning                                       |
| ------ | --------------------------------------------- |
| B → M  | If Bread is bought, Milk may also be bought   |
| M → B  | If Milk is bought, Bread may also be bought   |
| B → Bu | If Bread is bought, Butter may also be bought |
| Bu → B | If Butter is bought, Bread may also be bought |

# 🔹 **Step 1: Start from Frequent Itemsets**

---

## 👉 General Rule Format:

```text

```

👉 Meaning:

> If A is present, B is likely to be present

---

# 🔹 **Step 3: Rule Generation Logic**

---

## 💡 Important Rule:

> From one frequent itemset, we can generate multiple rules

---

---

## 🎯 Example 1: {Bread, Milk}

---

### Possible splits:

```text
Bread → Milk
Milk → Bread
```

---

👉 Why?

Because we split the set into:

* Left side (A)

---

## 🎯 Example 2: {Bread, Butter}

---

### Possible rules:

```text
Bread → Butter
Butter → Bread
```

---

---

# 🔹 **Step 4: General Rule Generation Formula**

---

## For itemset:

```text
{A, B}
```

👉 Possible rules:

```text
A → B
B → A
```

---

---

## For larger itemset:

```text
{A, B, C}
```

👉 Possible rules:

```text
| Rule     | Left Side (LHS) | Right Side (RHS) | Type          | Meaning (Simple Explanation)                |
| -------- | --------------- | ---------------- | ------------- | ------------------------------------------- |
| A → B, C | A               | B, C             | Single → Pair | If A is present, B and C are likely present |
| B → A, C | B               | A, C             | Single → Pair | If B is present, A and C are likely present |
| C → A, B | C               | A, B             | Single → Pair | If C is present, A and B are likely present |
| A, B → C | A, B            | C                | Pair → Single | If A and B are present, C is likely present |
| A, C → B | A, C            | B                | Pair → Single | If A and C are present, B is likely present |
| B, C → A | B, C            | A                | Pair → Single | If B and C are present, A is likely present |

```

---

---

# 🔹 **Step 5: Why Do We Generate Multiple Rules?**

---

Because:

👉 Relationship is **directional**

---

## 🎯 Example:

```text
Bread → Milk
```

means:

> If Bread is bought → Milk is likely

---

But:

```text
Milk → Bread
```

means:

> If Milk is bought → Bread is likely

---

👉 These are **different meanings**

---

---

# 🔹 **Step 6: Not All Rules Are Useful**

---

After generating rules, we check:

* Confidence
* Lift

---

## ❌ Example:

If:

```text
Confidence = very low
```

👉 Rule is **not useful**

# 🔹 Step 13: Calculate Confidence

## 📌 Formula

$$
Confidence(A \rightarrow B) =
\frac{Support(A \cap B)}{Support(A)}
$$

In simple words:

> Out of all transactions containing A, how many also contain B?

---

## Rule 1: Bread → Milk

Values:

$$
Support(B,M) = 0.50
$$

$$
Support(B) = 0.83
$$

Now:

$$
Confidence(B \rightarrow M) =
\frac{0.50}{0.83}
$$

$$
Confidence(B \rightarrow M) = 0.60
$$

# 🔹 **5. Limitations of Association Rule Mining**

---

# 📌 **5.1 Too Many Rules Problem**

---

## ❌ Problem

👉 Association Rule Mining can generate **hundreds or thousands of rules**

---

## 🎯 Example

From a dataset, we may get:

```text
Bread → Milk
Milk → Bread
Bread → Butter
Butter → Bread
Milk → Butter
Butter → Milk
...
```

👉 And many more combinations

---

## 🧠 Issue

👉 Not all rules are useful
👉 Hard to analyze everything

---

## 💡 Simple Understanding

> More data → more rules → more confusion

---

---

## ✅ Solution Idea (basic)

* Use higher support/confidence
* Filter only strong rules

---

---

# 📌 **5.2 Weak or Misleading Rules**

---

## ❌ Problem

👉 Some rules look strong but are **not meaningful**

---

## 🎯 Example

```text
Milk → Bread
Confidence = 0.75
```

👉 Looks strong

BUT:

* Bread is already very common

---

## 🧠 Issue

👉 Rule may happen **by chance**, not real relationship

---

---

## 💡 Important Insight

👉 This is why we use **Lift**

---

## Rule:

* Lift > 1 → strong
* Lift = 1 → no relation
* Lift < 1 → weak

---

---

# 📌 **5.3 Sensitivity to Support/Confidence Thresholds**

---

## ❌ Problem

👉 Results depend heavily on chosen thresholds

---

## 🎯 Example

### Case 1:

```text
Min Support = 0.50
```

👉 Only few rules

---

### Case 2:

```text
Min Support = 0.20
```

👉 Many rules

---

---

## 🧠 Issue

👉 Wrong threshold → wrong conclusions

---

---

## 💡 Simple Understanding

> Too high → miss important patterns
> Too low → too many useless rules

---

---

# 📌 **5.4 Computational Complexity**

---

## ❌ Problem

👉 Apriori generates **many combinations**

---

## 🎯 Example

If we have 100 items:

👉 Possible combinations:

```text
Very large (exponential growth)
```

---

---

## 🧠 Issue

👉 Algorithm becomes:

* slow
* resource-intensive

---

---

## 💡 Simple Understanding

> More items → more combinations → more computation

---

---

## ✅ Real-world note

👉 That’s why faster algorithms exist (e.g., FP-Growth)

````python
# ============================================================
# WEEK 14: ASSOCIATION RULE MINING USING APRIORI
# Complete Beginner-Friendly Code
# ============================================================

# If mlxtend is not installed in Colab, run this first:
# !pip install mlxtend

import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# ------------------------------------------------------------
# STEP 1: Create Transaction Dataset
# ------------------------------------------------------------

transactions = [
    ["Bread", "Milk"],
    ["Bread", "Butter"],
    ["Milk", "Butter"],
    ["Bread", "Milk", "Butter"],
    ["Bread", "Milk"],
    ["Bread", "Butter"]
]

print("=" * 70)
print("STEP 1: TRANSACTION DATA")
print("=" * 70)

for i, transaction in enumerate(transactions, start=1):
    print(f"T{i}: {transaction}")

# ------------------------------------------------------------
# STEP 2: Convert Transaction Data into Binary Format
# ------------------------------------------------------------

encoder = TransactionEncoder()
encoded_array = encoder.fit(transactions).transform(transactions)

df = pd.DataFrame(encoded_array, columns=encoder.columns_)

print("\n" + "=" * 70)
print("STEP 2: BINARY TRANSACTION TABLE")
print("=" * 70)

print(df.astype(int))

# ------------------------------------------------------------
# STEP 3: Find Frequent Itemsets using Apriori
# ------------------------------------------------------------

min_support_value = 0.50

frequent_itemsets = apriori(
    df,
    min_support=min_support_value,
    use_colnames=True
)

print("\n" + "=" * 70)
print("STEP 3: FREQUENT ITEMSETS")
print("=" * 70)

print(f"Minimum Support = {min_support_value}")
print(frequent_itemsets)

# ------------------------------------------------------------
# STEP 4: Add Itemset Length
# ------------------------------------------------------------

frequent_itemsets["length"] = frequent_itemsets["itemsets"].apply(lambda x: len(x))

print("\n" + "=" * 70)
print("STEP 4: FREQUENT ITEMSETS WITH LENGTH")
print("=" * 70)

print(frequent_itemsets)

# ------------------------------------------------------------
# STEP 5: Generate Association Rules
# ------------------------------------------------------------

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.50
)

print("\n" + "=" * 70)
print("STEP 5: ASSOCIATION RULES")
print("=" * 70)

rules_display = rules[
    ["antecedents", "consequents", "support", "confidence", "lift"]
]

print(rules_display)

# ------------------------------------------------------------
# STEP 6: Make Rules Easier to Read
# ------------------------------------------------------------

def convert_to_string(itemset):
    return ", ".join(list(itemset))

rules["Antecedent"] = rules["antecedents"].apply(convert_to_string)
rules["Consequent"] = rules["consequents"].apply(convert_to_string)

final_rules = rules[
    ["Antecedent", "Consequent", "support", "confidence", "lift"]
]

print("\n" + "=" * 70)
print("STEP 6: CLEAN RULE TABLE")
print("=" * 70)

print(final_rules)

# ------------------------------------------------------------
# STEP 7: Interpret Each Rule
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STEP 7: RULE INTERPRETATION")
print("=" * 70)

for index, row in final_rules.iterrows():
    antecedent = row["Antecedent"]
    consequent = row["Consequent"]
    support = row["support"]
    confidence = row["confidence"]
    lift = row["lift"]

    print(f"\nRule: {antecedent} → {consequent}")
    print(f"Support    = {support:.2f}")
    print(f"Confidence = {confidence:.2f}")
    print(f"Lift       = {lift:.2f}")

    print(
        f"Meaning: When a customer buys {antecedent}, "
        f"there is a {confidence * 100:.0f}% chance they also buy {consequent}."
    )

    if lift > 1:
        print("Lift Interpretation: Strong positive relationship.")
    elif lift == 1:
        print("Lift Interpretation: No special relationship.")
    else:
        print("Lift Interpretation: Weak or negative relationship.")

# ------------------------------------------------------------
# STEP 8: Visualize Item Frequency
# ------------------------------------------------------------

item_counts = df.sum().sort_values(ascending=False)

plt.figure(figsize=(7, 5))
bars = plt.bar(item_counts.index, item_counts.values)

plt.title("Item Frequency in Transactions")
plt.xlabel("Items")
plt.ylabel("Count")

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.05,
        str(int(height)),
        ha="center"
    )

plt.ylim(0, max(item_counts.values) + 1)
plt.grid(axis="y")
plt.show()

# ------------------------------------------------------------
# STEP 9: Visualize Support of Frequent Itemsets
# ------------------------------------------------------------

frequent_itemsets["itemsets_str"] = frequent_itemsets["itemsets"].apply(convert_to_string)

plt.figure(figsize=(8, 5))
bars = plt.bar(
    frequent_itemsets["itemsets_str"],
    frequent_itemsets["support"]
)

plt.title("Support of Frequent Itemsets")
plt.xlabel("Itemsets")
plt.ylabel("Support")
plt.xticks(rotation=30)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.02,
        f"{height:.2f}",
        ha="center"
    )

plt.ylim(0, 1.1)
plt.grid(axis="y")
plt.show()

# ------------------------------------------------------------
# STEP 10: Visualize Confidence of Rules
# ------------------------------------------------------------

final_rules = final_rules.copy()
final_rules.loc[:, "Rule"] = final_rules["Antecedent"] + " → " + final_rules["Consequent"]

plt.figure(figsize=(8, 5))
bars = plt.bar(final_rules["Rule"], final_rules["confidence"])

plt.title("Confidence of Association Rules")
plt.xlabel("Rules")
plt.ylabel("Confidence")
plt.xticks(rotation=30)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.02,
        f"{height:.2f}",
        ha="center"
    )

plt.ylim(0, 1.1)
plt.grid(axis="y")
plt.show()

# ------------------------------------------------------------
# STEP 11: Visualize Lift of Rules
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))
bars = plt.bar(final_rules["Rule"], final_rules["lift"])

plt.title("Lift of Association Rules")
plt.xlabel("Rules")
plt.ylabel("Lift")
plt.axhline(y=1, linestyle="--", label="Lift = 1")
plt.xticks(rotation=30)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.02,
        f"{height:.2f}",
        ha="center"
    )

plt.ylim(0, max(final_rules["lift"]) + 0.4)
plt.legend()
plt.grid(axis="y")
plt.show()

# ------------------------------------------------------------
# STEP 12: Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("1. We created transaction data.")
print("2. We converted it into binary format.")
print("3. We used Apriori to find frequent itemsets.")
print("4. We generated association rules.")
print("5. We evaluated rules using Support, Confidence, and Lift.")

print("\nImportant Concepts:")
print("Support    = How common the itemset is.")
print("Confidence = How often the rule is correct.")
print("Lift       = How strong the relationship is compared to chance.")
````
